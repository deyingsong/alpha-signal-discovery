# Chapter 26: Exotic Options

> **Color Convention**  
> 🟡 **Yellow** = Definition / Theorem / Property  
> 🟢 **Green** = Comment / Application / Insight

---

## Overview

Exotic options are nonstandard OTC derivatives engineered for specific hedging, regulatory, or speculative needs. They are more profitable for dealers than vanilla options but require sophisticated valuation.

---

## 26.1 Packages

🟡 **Definition:**

> A **package** = portfolio of European calls/puts, forwards, cash, and the underlying. Examples: bull/bear/butterfly spreads, straddles, strangles, range forwards.

🟡 **Property:**

> Any derivative can be converted to a **zero-cost product** by deferring payment to maturity $T$. If the upfront cost is $c$, the deferred cost is $c\,e^{rT}$.

---

## 26.2 Perpetual American Options

Closed-form for perpetual American call (infinite maturity):

$$c = S_0\left(\frac{H}{S_0}\right)^{2r/\sigma^2} \quad \text{when } S_0 < H$$

where the optimal exercise barrier is:

$$H = K \cdot \frac{2r/\sigma^2}{2r/\sigma^2 - 1}$$

---

## 26.3 Nonstandard American Options

🟢 **Application:**

> OTC American options often have nonstandard features:
> - **Bermudan options**: Exercise only on specified dates (between American and European)
> - **Lock-out period**: No exercise allowed until a future date
> - **Step-up strike**: Strike price increases over time

---

## 26.4 Gap Options

🟡 **Definition:**

> A **gap call** pays $S_T - K_1$ when $S_T > K_2$ (where $K_1$ may differ from $K_2$):
>
> $$c_{\text{gap}} = S_0 e^{-qT} N(d_1) - K_1 e^{-rT} N(d_2)$$
>
> (Same as BSM but with $K_1$ instead of $K$ in the second term, using $K_2$ to compute $d_1$, $d_2$.)
>
> Gap call = regular call with strike $K_2$ + $(K_2 - K_1) \times$ asset-or-nothing binary call.

---

## 26.5 Forward Start Options

🟡 **Definition:**

> A **forward start option** commences at future time $t_1$ with the strike set to the then-prevailing asset price (at-the-money start).
>
> Value at time 0 = standard ATM BSM formula evaluated with time to maturity $T - t_1$ (since an ATM option's value scales with $S_0$, the current price cancels).

---

## 26.6 Cliquet Options

🟡 **Definition:**

> A **cliquet** (ratchet) = series of forward-start options. At each reset date $t_i$, the strike is reset to the current price $S_{t_i}$. The total payoff = sum of individual option payoffs.

---

## 26.7 Compound Options

🟡 **Definition:**

> **Compound options** = options on options. Four types:
> - Call on call, put on call, call on put, put on put
>
> Two critical dates: $T_1$ (expiry of the outer option) and $T_2 > T_1$ (expiry of the inner option). Analytical BSM-type formulas exist using the bivariate normal distribution.

---

## 26.8 Chooser Options

🟡 **Definition:**

> A **chooser** allows the holder at time $T_1$ to choose whether the option is a call or put (maturity $T_2$, strike $K$).
>
> At $T_1$: value = $\max(c_{T_1}, p_{T_1})$.
>
> Using put-call parity, a chooser decomposes into:
> - A call with strike $K$, maturity $T_2$
> - A put with strike $Ke^{-r(T_2-T_1)}$, maturity $T_1$

---

## 26.9 Barrier Options

🟡 **Definition:**

> **Barrier options**: payoff depends on whether $S$ reaches barrier $H$ during the option life.
>
> **Knock-out**: ceases to exist if $S$ hits $H$  
> **Knock-in**: comes into existence when $S$ hits $H$

🟡 **Key Identity:**

> $$\text{knock-in} + \text{knock-out} = \text{regular option}$$

🟡 **Eight Types:**

> | Direction | Up | Down |
> |-----------|-----|------|
> | Knock-out call | Up-and-out call ($H > S_0$) | Down-and-out call ($H < S_0$) |
> | Knock-in call | Up-and-in call ($H > S_0$) | Down-and-in call ($H < S_0$) |
> | Knock-out put | Up-and-out put ($H > S_0$) | Down-and-out put ($H < S_0$) |
> | Knock-in put | Up-and-in put ($H > S_0$) | Down-and-in put ($H < S_0$) |
>
> BSM-type closed-form formulas exist for all eight types.

🟡 **Definition (Parisian Option):**

> A **Parisian option** requires $S$ to spend a specified continuous period above/below the barrier before knock-in/out occurs. More path-dependent and harder to value than standard barrier options.

---

## 26.10 Binary Options

🟡 **Definition:**

> **Binary (digital) options** have discontinuous payoffs:
>
> | Type | Payoff | Formula |
> |------|--------|---------|
> | Cash-or-nothing call | $Q$ if $S_T > K$; else 0 | $Qe^{-rT}N(d_2)$ |
> | Cash-or-nothing put | $Q$ if $S_T < K$; else 0 | $Qe^{-rT}N(-d_2)$ |
> | Asset-or-nothing call | $S_T$ if $S_T > K$; else 0 | $S_0 e^{-qT}N(d_1)$ |
> | Asset-or-nothing put | $S_T$ if $S_T < K$; else 0 | $S_0 e^{-qT}N(-d_1)$ |
>
> **Decomposition**: Regular call = asset-or-nothing call $-$ $K e^{-rT}$ $\times$ cash-or-nothing call.

---

## 26.11 Lookback Options

🟡 **Definition:**

> **Lookback options**: payoff depends on the maximum or minimum of $S$ over the option life.
>
> - **Floating lookback call**: pays $S_T - S_{\min}$ (never out-of-the-money)
> - **Floating lookback put**: pays $S_{\max} - S_T$
> - **Fixed lookback call**: pays $\max(0, S_{\max} - K)$
> - **Fixed lookback put**: pays $\max(0, K - S_{\min})$
>
> Closed-form BSM-type formulas exist. Lookbacks are expensive — they guarantee the best possible exercise price in hindsight.

---

## 26.12 Shout Options

🟡 **Definition:**

> A **shout option**: the holder can "shout" once during the option's life, locking in the intrinsic value at that moment. At expiry:
>
> $$\text{Payoff} = \max(\text{regular payoff}, S_\tau - K) = (S_\tau - K) + \max(0, S_T - S_\tau)$$
>
> where $\tau$ = shout time. Equivalent to: intrinsic value at shout + a new ATM call starting at shout time.

---

## 26.13 Asian Options

🟡 **Definition:**

> **Asian options**: payoff depends on the average asset price $S_{avg}$:
>
> - **Average price call**: $\max(0, S_{avg} - K)$
> - **Average price put**: $\max(0, K - S_{avg})$
> - **Average strike call**: $\max(0, S_T - S_{avg})$
> - **Average strike put**: $\max(0, S_{avg} - S_T)$
>
> Arithmetic average: no closed form → use Monte Carlo (geometric average as control variate). Geometric average: closed-form BSM-type formula.

---

## 26.14 Exchange Options

🟡 **Definition:**

> **Exchange option** (Margrabe, 1978): right to exchange asset $B$ for asset $A$:
>
> $$c = S_A e^{-q_A T} N(d_1) - S_B e^{-q_B T} N(d_2)$$
>
> where $\hat{\sigma}^2 = \sigma_A^2 + \sigma_B^2 - 2\rho\sigma_A\sigma_B$ and $d_1$, $d_2$ use $\hat{\sigma}$.

---

## 26.15 Rainbow Options

🟡 **Definition:**

> **Rainbow options** depend on two or more risky assets:
> - **Best-of**: $\max(S_1, S_2)$ — generally requires Monte Carlo
> - **Basket option**: payoff on a weighted portfolio of assets

---

## 26.16 Volatility and Variance Swaps

🟡 **Definition (Volatility Swap):**

> Exchange realized volatility $\bar{\sigma}$ for a fixed rate $K_{vol}$:
>
> $$\text{Payoff} = N_{vega}(\bar{\sigma} - K_{vol})$$

🟡 **Definition (Variance Swap):**

> Exchange realized variance $\bar{\sigma}^2$ for a fixed variance $K_{var}$:
>
> $$\text{Payoff} = N_{var}(\bar{\sigma}^2 - K_{var})$$
>
> **Replication**: A variance swap can be replicated by a static portfolio of out-of-the-money puts and calls across all strikes:
>
> $$K_{var} = \frac{2}{T}\left[\int_0^{F_0}\frac{p(K)}{K^2}dK + \int_{F_0}^{\infty}\frac{c(K)}{K^2}dK\right]$$
>
> This is the basis for the VIX volatility index calculation.

---

## 26.17 Static Options Replication

Replicate an exotic option with a **static portfolio** of vanilla options matched at the exotic's boundary. No rebalancing needed until the boundary is reached. More robust to model error than dynamic hedging.

---

## Summary

| Exotic Type | Key Characteristic |
|-------------|-------------------|
| Barrier | Knock-in/out at level $H$ |
| Binary/Digital | Discontinuous payoff |
| Lookback | Payoff on realized max/min of $S$ |
| Asian | Payoff on average of $S$ |
| Shout | Holder locks in intrinsic value once |
| Chooser | Holder picks call or put at $T_1$ |
| Compound | Option on option |
| Exchange | Right to swap one asset for another |
| Variance swap | Exchange realized variance for fixed rate |
| Cliquet | Series of forward-start reset options |
