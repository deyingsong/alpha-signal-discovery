## Overview

**VaR** and **ES** are single-number risk measures that summarize total portfolio risk. Used by banks, fund managers, and regulators worldwide.

---

## 22.1 The VaR and ES Measures

### Value at Risk (VaR)

"I am $X$% certain there will not be a loss of more than $V$ dollars over the next $N$ days."

- $V$ = VaR at confidence level $X$%, horizon $N$ days
- Basel standard: $N = 10$, $X = 99\%$

🟢 **Comment:**

> VaR is attractive because it is easy to understand — it answers the core management question: *"How bad can things get?"* It compresses hundreds of Greek-letter risk measures into a single number that executives can grasp.
>
> **Weakness**: VaR says nothing about losses *beyond* the threshold. A portfolio can be structured to have a favorable VaR while hiding extreme tail losses.

### Expected Shortfall (ES)

The **expected loss conditional on exceeding VaR**:

$$\text{ES} = E[\text{Loss} \mid \text{Loss} > \text{VaR}]$$

ES is a **coherent risk measure** (satisfies subadditivity: $\text{ES}(A+B) \leq \text{ES}(A) + \text{ES}(B)$). Bank regulators are moving from VaR to ES for market risk capital.

---

## 22.2 Historical Simulation

**Algorithm:**
1. Collect daily percentage changes in all market variables over past $m$ days (typically 500–1000)
2. Apply each historical day's changes to today's portfolio → $m$ scenarios
3. Sort simulated portfolio P&L
4. **VaR** = $(1-X\%)$th percentile loss; **ES** = mean of losses beyond VaR

**Advantages:** No distributional assumption; captures fat tails and non-linear correlations naturally.  
**Disadvantages:** Limited to the historical window; slow to respond to regime changes.

---

## 22.3 Model-Building Approach

Assume daily changes in portfolio value are normally distributed.

For a single position with dollar exposure $w$ and daily standard deviation $\sigma$:

$$\text{VaR}_{1\text{-day}, X\%} = w\sigma \cdot z_X$$

$$\text{VaR}_{N\text{-day}} = \sqrt{N} \cdot \text{VaR}_{1\text{-day}}$$

For ES under normality:

$$\text{ES} = w\sigma \cdot \frac{\phi(z_X)}{1-X}$$

where $\phi$ is the standard normal PDF.

---

## 22.4 The Linear Model

For a portfolio of assets with dollar exposures $x_i$:

$$\sigma_P^2 = \sum_i \sum_j x_i x_j \sigma_i \sigma_j \rho_{ij} = \mathbf{x}^T \mathbf{\Sigma} \mathbf{x}$$

where $\mathbf{\Sigma}$ is the variance-covariance matrix of daily returns.

🟢 **Comment:**

> The linear model assumes portfolio P&L is a linear combination of market variable changes. This holds exactly for equities and FX positions, but **not for options** (which have nonlinear payoffs). The model is fast and tractable but inaccurate for options portfolios.

---

## 22.5 The Quadratic Model (Delta-Gamma)

For portfolios with options:

$$\Delta P \approx \sum_i \delta_i \Delta x_i + \frac{1}{2}\sum_i \gamma_i (\Delta x_i)^2$$

This leads to a **non-normal distribution** for $\Delta P$. The distribution can be derived analytically (using moment matching) or estimated via Monte Carlo.

---

## 22.6 Monte Carlo Simulation for VaR

🟢 **Application:**

> Monte Carlo VaR procedure:
> 1. Value the portfolio today
> 2. Sample correlated changes in all market variables from their assumed joint distribution
> 3. Revalue the portfolio under each scenario
> 4. Repeat thousands of times; find the relevant loss percentile

Handles non-linearity perfectly but is computationally intensive.

---

## 22.7 Comparison of Approaches

🟢 **Comment:**

> | Method | Nonlinearity | Speed | Fat tails |
> |--------|-------------|-------|-----------|
> | Historical simulation | Yes | Fast | Yes (implicitly) |
> | Linear model | No | Very fast | No (assumes normal) |
> | Monte Carlo | Yes | Slow | Depends on assumptions |
>
> Most large banks use **historical simulation** as the primary approach, supplemented by stress testing.

---

## 22.8 Back Testing

Compare VaR forecasts to actual P&L. At 99% confidence, we expect ~2.5 "exceptions" (loss > VaR) per year. Basel II specifies capital add-ons based on the number of exceptions in the last 250 trading days:

| Exceptions | Zone | Capital Add-on |
|------------|------|----------------|
| 0–4 | Green | None |
| 5–9 | Yellow | Increasing |
| 10+ | Red | Maximum |

---

## 22.9 Principal Components Analysis (PCA)

PCA identifies orthogonal factors explaining most of the variance in yield curves / return correlations. The first PC explains the largest fraction of variance (e.g., parallel shifts for interest rates). This reduces dimensionality dramatically for VaR calculations involving many correlated market variables.

## Summary

| Concept | Key Formula / Fact |
|---------|-------------------|
| VaR at $X$% | $(1-X\%)$ quantile of loss distribution |
| ES | Mean loss beyond VaR; coherent risk measure |
| N-day VaR scaling | $\text{VaR}_N = \sqrt{N} \cdot \text{VaR}_1$ (under normality) |
| Historical simulation | Uses past scenarios directly; no distributional assumption |
| Linear model | Normal distribution assumption; fast |
| Monte Carlo | Flexible; handles nonlinearity |
