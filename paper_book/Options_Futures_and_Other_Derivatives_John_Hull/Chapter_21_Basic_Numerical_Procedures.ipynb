## Overview

Three numerical methods for valuing derivatives without closed-form solutions:

1. **Binomial Trees** — best for American options with early-exercise features
2. **Monte Carlo Simulation** — best for path-dependent or multi-asset derivatives  
3. **Finite Difference Methods** — grid-based PDE solvers, handles American options

---

## 21.1 Binomial Trees

🟢 **Application (Risk-Neutral Valuation):**

> A derivative can be valued assuming the world is risk-neutral:
> 1. Assume all assets earn the risk-free rate $r$
> 2. Calculate the expected payoff of the derivative
> 3. Discount at the risk-free rate

🟡 **Property (Cox–Ross–Rubinstein Parameters):**

> The change in $\ln S$ over $\Delta t$ has variance $\sigma^2 \Delta t$. CRR parameterization:
>
> $$u = e^{\sigma\sqrt{\Delta t}}, \quad d = e^{-\sigma\sqrt{\Delta t}} = 1/u$$
>
> $$p = \frac{e^{(r-q)\Delta t} - d}{u - d}$$
>
> For a European option, as the number of steps $n \to \infty$, the binomial price converges to BSM.

### Valuing American Options

At each node: $V = \max(\text{continuation value}, \text{intrinsic value})$.

Backward induction: start from terminal payoffs, discount backward applying the early-exercise check at each node.

## 21.2 Binomial Trees for Indices, Currencies, and Futures

| Underlying | Risk-neutral drift in $p$ |
|------------|--------------------------|
| Stock index | $r - q$ (dividend yield $q$) |
| Currency | $r - r_f$ (foreign risk-free rate) |
| Futures | $0$ (futures price is a martingale) |

## 21.3 Dividend-Paying Stocks

**Known discrete dividend $D$**: At the ex-dividend date, the tree branches, and the price drops by $D$.

**Escrow method**: Separate the stock into a risk-free (present value of future dividends) and a risky component. Build the tree on the risky component only.

---

## 21.4 Alternative Tree Procedures

- **Trinomial trees**: Three branches per node (more flexible than binomial)
- **Non-recombining trees**: Useful but computationally expensive — node count grows as $2^n$

---

## 21.5 Time-Dependent Parameters

Parameters ($r$, $\sigma$, $q$) can be made functions of time to match the observed term structure.

## 21.6 Monte Carlo Simulation

**Procedure (risk-neutral):**

1. Simulate a path: $\Delta S = (r-q)S\Delta t + \sigma S\epsilon\sqrt{\Delta t}$, $\epsilon \sim N(0,1)$
2. Compute the derivative payoff at maturity
3. Repeat $N$ times, average the discounted payoffs

**Standard error** $\approx \hat{\omega}/\sqrt{N}$ where $\hat{\omega}$ = std dev of simulated payoffs.

🟢 **Application:**

> Monte Carlo is well-suited to **path-dependent** derivatives (Asian, lookback options) and **multi-asset** options. However, it is not directly applicable to American options, which require backward induction.

🟢 **Application (Derivative Pricing):**

> For any derivative with payoff at time $T$, regardless of path complexity, the Monte Carlo approach works seamlessly by simply computing the payoff along each simulated path and averaging.

## 21.7 Variance Reduction Procedures

| Technique | Description |
|-----------|-------------|
| **Antithetic variable** | For each path with $\{\epsilon_i\}$, also simulate $\{-\epsilon_i\}$; average the two payoffs |
| **Control variate** | Use a related option with known BSM price to reduce error |
| **Importance sampling** | Oversample from the important region of the distribution |
| **Stratified sampling** | Divide probability space into strata, sample proportionally |
| **Quasi-random (low-discrepancy)** | Use Sobol or Halton sequences instead of pseudorandom numbers |

---

## 21.8 Finite Difference Methods

Discretize the BSM PDE on a $(S, t)$ grid and solve numerically.

**Explicit method**: Forward in time — simple but conditionally stable.

**Implicit method**: Backward in time — requires solving a tridiagonal system; always stable.

**Crank–Nicolson**: Average of explicit and implicit — second-order accuracy, unconditionally stable.

| Method | Stability | Order of Accuracy |
|--------|-----------|-------------------|
| Explicit | Conditional | $O(\Delta t, \Delta S^2)$ |
| Implicit | Unconditional | $O(\Delta t, \Delta S^2)$ |
| Crank–Nicolson | Unconditional | $O(\Delta t^2, \Delta S^2)$ |

**Key link**: The explicit finite difference method is mathematically equivalent to a trinomial tree.

## Summary

| Method | Best For | Handles Early Exercise? |
|--------|----------|------------------------|
| Binomial tree | American options | Yes |
| Monte Carlo | Path-dependent, multi-asset | No (without modification) |
| Finite difference | American options, PDEs | Yes |
