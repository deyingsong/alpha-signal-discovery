# Chapter 10: Mechanics of Options Markets

> **Color Convention**  
> 🟡 **Yellow** = Definition / Theorem / Property  
> 🟢 **Green** = Comment / Application / Insight

## Overview

This chapter explains the mechanics of options markets: types of options, exercise conventions, moneyness, exchange-traded vs. OTC options, margin requirements, and the distinction between exchange-traded options (calls, puts) and other option-like instruments (warrants, employee stock options, convertibles).

---

## 10.1 Types of Options

🟡 **Definition:**

> A **call option** gives the holder the right (not obligation) to **buy** an asset by a certain date for a certain price.
>
> A **put option** gives the holder the right (not obligation) to **sell** an asset by a certain date for a certain price.
>
> The date specified in the contract is the **expiration date** (maturity). The price specified is the **exercise price** (strike price).

🟡 **Definition (American vs. European):**

> **American options** can be exercised at **any time** up to and including the expiration date.  
> **European options** can be exercised **only at** the expiration date.
>
> (The distinction has nothing to do with geographic location.)

---

## 10.2 Option Positions

🟡 **Property:**

> There are two sides to every option contract:
> - **Long position** (buyer): pays the premium; has the right to exercise
> - **Short position** (writer/seller): receives the premium upfront; has the **obligation** to fulfill the contract if exercised
>
> The writer's potential loss is theoretically unlimited (for a naked call) — this is why margin is required.

---

## 10.3 Underlying Assets

Options exist on:
- Individual stocks (most common in examples)
- Stock indices (e.g., S&P 500 options)
- Foreign currencies
- Futures contracts
- ETFs

---

## 10.4 Specification of Exchange-Traded Options

🟡 **Definition (Expiration Date):**

> For U.S. exchange-traded stock options, the expiration date is the **third Friday** of the expiration month.

🟡 **Definition (Option Class and Series):**

> All options of the **same type** (calls or puts) on a stock = an **option class**. An **option series** = all contracts of the same type with the same strike and expiration. If options trade at every expiration and every strike, with calls and puts at each, there are $2 \times \text{strikes} \times \text{expirations}$ different series.

🟡 **Definition (Moneyness):**

> For a stock price $S$ and strike $K$:
>
> | Status | Call | Put |
> |--------|------|-----|
> | In the money | $S > K$ | $S < K$ |
> | At the money | $S = K$ | $S = K$ |
> | Out of the money | $S < K$ | $S > K$ |

🟡 **Definition (Intrinsic Value):**

> The **intrinsic value** of an option = the value if exercised immediately:
> - Call: $\max(S - K, 0)$
> - Put: $\max(K - S, 0)$
>
> The **time value** = option price − intrinsic value. American options always have non-negative time value before expiration.

---

## 10.5 Trading

🟡 **Definition (Position and Exercise Limits):**

> Exchanges specify a **position limit** = maximum number of option contracts an investor can hold on one side of the market (long calls + short puts = one side; short calls + long puts = the other side).
>
> The **exercise limit** (usually equal to the position limit) = maximum number of contracts that can be exercised within any 5 consecutive business days. Most actively traded options have larger limits.

---

## 10.6 Margin Requirements

🟡 **Property:**

> An option **buyer** pays the premium in full upfront — no margin required.
>
> An option **writer** must post margin because of potential future obligations. For a **naked call** (writer doesn't hold the underlying), the CBOE requires the greater of:
> 1. 25% of underlying value + premium − out-of-the-money amount
> 2. 10% of underlying value + premium

Buying on margin (borrowing from broker) for stock purchases: in the U.S., investors can borrow up to 50% of the purchase price (Regulation T).

---

## 10.7 Options on Other Instruments

🟡 **Definition (Warrants):**

> **Warrants** are options issued by a financial institution or corporation. A financial institution might issue put warrants on gold or stock index warrants. When a company issues warrants on its own stock, exercise dilutes existing shareholders.

🟡 **Definition (Employee Stock Options):**

> **Employee stock options (ESOs)** are call options issued by a company to employees — typically at-the-money at issuance, with a vesting period. Accounting standards require them to be **expensed at fair value** on the income statement.

🟡 **Definition (Convertible Bonds):**

> **Convertible bonds** are bonds with an embedded option allowing the holder to convert the bond into equity at a specified **conversion ratio**. Value = straight bond value + value of embedded call option on the company's stock.

🟡 **Definition (OTC Options):**

> OTC options are customized — any strike, maturity, and contract size are possible. They require counterparties to **post collateral** (under CSAs) to reduce credit risk. OTC options include interest rate caps, floors, swaptions, and exotic options.

## Summary

| Concept | Key Point |
|---------|-----------|
| Call option | Right to buy at strike $K$ |
| Put option | Right to sell at strike $K$ |
| American option | Can be exercised any time before expiry |
| European option | Can only be exercised at expiry |
| Intrinsic value | $\max(S-K, 0)$ for calls; $\max(K-S, 0)$ for puts |
| Writer's margin | Required because writer has potential obligations |
| Convertible bond | Bond + embedded call option on equity |
