## Overview

This chapter covers interest rate swaps, currency swaps, and other swap variants. It explains how swaps work, how they are valued, and how credit risk and market risk arise in swap contracts.

---

## 7.1 Mechanics of Interest Rate Swaps

🟡 **Definition:**

> An **interest rate swap** is an agreement to exchange interest at a predetermined **fixed rate** for interest at a **floating reference rate** (e.g., overnight SOFR) on the same notional principal for an agreed period. No exchange of principal occurs.
>
> The **OIS rate** = the fixed rate in an overnight index swap (OIS), where the floating leg compounds daily overnight rates. OIS swaps have zero initial value.

🟡 **Definition (Bid/Ask/Swap Rate):**

> A swap market maker quotes:
> - **Bid** = fixed rate paid when receiving floating
> - **Ask** = fixed rate received when paying floating
> - **Swap rate** = mid-point (bid + ask) / 2
>
> Swap rates for various maturities form the **swap rate curve** — the backbone of the fixed-income market.

🟢 **Comment (LIBOR vs. OIS):**

> LIBOR rates are **estimated** interbank borrowing rates for 1-day to 12-month periods. OIS rates are based on **actual** overnight transactions — nearly risk-free, collateralized. Key differences:
> - LIBOR includes bank credit and liquidity risk
> - OIS is nearly risk-free (secured overnight lending)
> - The **LIBOR–OIS spread** widens during financial stress (e.g., 2008 crisis)
>
> LIBOR is being replaced by overnight rates (SOFR, SONIA, €STR) globally.

🟢 **Application (OIS Example):**

> A 2-year OIS: Apple pays 3% fixed (quarterly) to Citigroup; Citigroup pays the compounded overnight SOFR rate on $100M. Only net payments are exchanged. Each period, Apple pays if SOFR < 3%, Citigroup pays if SOFR > 3%.

## 7.2 Valuation of Interest Rate Swaps

🟢 **Application:**

> Each exchange of payments in a swap is a **Forward Rate Agreement (FRA)**. An interest rate swap = a **portfolio of FRAs**.
>
> Value the swap by assuming forward rates are realized:
>
> $$V_{\text{swap}} = \sum_{i=1}^{n} e^{-r_i t_i} (R_F^i - R_K)\,L\,\delta_i$$
>
> where $R_F^i$ = forward rate for period $i$, $R_K$ = fixed swap rate, $L$ = notional, $\delta_i$ = accrual fraction.

Alternatively, value as a **bond portfolio**:
- **Fixed-rate payer**: long floating bond, short fixed coupon bond
- **Fixed-rate receiver**: long fixed coupon bond, short floating bond

## 7.3 Currency Swaps

🟡 **Definition:**

> A **fixed-for-fixed currency swap** exchanges principal + fixed interest in one currency for principal + fixed interest in another currency. Unlike interest rate swaps, **principal IS exchanged** — at initiation and at maturity (at the same exchange rate).

🟡 **Property:**

> Two additional types:
> 1. **Fixed-for-floating**: exchange fixed interest in one currency for floating in another
> 2. **Floating-for-floating**: exchange floating in one currency for floating in another (a basis swap across currencies)

---

## 7.4 Credit Risk in Swaps

🟡 **Property:**

> For a financial institution in a swap, two risks exist:
> - **Credit risk**: potential loss if the counterparty defaults when the swap has positive value to the institution
> - **Market risk**: risk that the swap's value changes due to market movements (rates, FX)
>
> Credit risk in a swap is much less than in a loan — only the net present value is at risk, not the entire notional principal.

🟡 **Definition (CDS):**

> A **Credit Default Swap (CDS)** provides insurance against default by a reference entity. The buyer pays a periodic premium (the CDS spread) to the seller; the seller pays the loss given default if a credit event occurs.

## 7.5 Other Swap Variants

🟡 **Definition:**

> **Amortizing swap**: notional principal decreases over the swap life (matches a loan amortization schedule).  
> **Step-up swap**: notional principal increases over the swap life.

🟡 **Definition:**

> **Compounding swap**: instead of periodic interest payments, interest is compounded to the end of the swap life, with one final net payment.

🟡 **Definition:**

> **Accrual swap**: interest accrues only on days when the floating reference rate is within a specified range (e.g., above a floor and below a cap).

🟡 **Definition:**

> **Equity swap**: one party pays the **total return** on an equity index (dividends + capital gains) on a notional; the other pays fixed or floating interest.

🟡 **Definition (Swaptions):**

> A **swaption** is an option to enter into an interest rate swap at a future date. An **extendable swap** gives one party the right to extend the swap. A **puttable swap** gives one party the right to terminate early. All embed swaption-like options.

🟡 **Definition:**

> **Commodity swap** = series of forward contracts on a commodity with different maturities but the same delivery price. One party pays fixed; the other pays market price.

🟡 **Definition:**

> **Volatility swap**: one party pays a pre-agreed volatility, the other pays realized historical volatility, for a series of time periods. The payoff depends on whether realized volatility exceeds or falls below the agreed level.

🟡 **Definition (Diff/Quanto Swap):**

> A **differential swap (diff swap / quanto swap)**: a floating rate observed in one currency is applied to a principal in another currency. Requires a **quanto adjustment** (Chapter 30) for the correlation between the rate and the exchange rate.

## Summary

| Swap Type | Description |
|-----------|-------------|
| Interest rate swap | Fixed vs. floating on same notional |
| OIS | Fixed vs. compounded overnight rate |
| Currency swap | Fixed in one currency vs. fixed in another; principal exchanged |
| Equity swap | Index return vs. fixed or floating |
| Amortizing / step-up | Variable notional |
| Compounding swap | Interest compounded to maturity |
| CDS | Insurance against default |
| Swaption | Option to enter a swap |
| Commodity swap | Fixed vs. market commodity price |
