# Chapter 9: XVAs

## Overview

XVA (X-Valuation Adjustment) refers to a family of adjustments to the "clean" price of a derivative, accounting for credit risk, funding costs, margin costs, and capital requirements. The key XVAs are CVA, DVA, FVA, MVA, and KVA.

---

## 9.1 Netting

🟡 **Definition:**

> Under a **netting agreement** (ISDA Master Agreement), if a counterparty defaults, all outstanding derivatives between the two parties are immediately terminated and a single net settlement amount is calculated — the bank either owes or is owed the **net** value, not the sum of gross positive values.
>
> Netting dramatically reduces credit exposure: if a bank has 10 trades with a counterparty worth $+5M$ each and 10 worth $-4M$ each, without netting exposure is $50M$; with netting it is $10M$.

---

## 9.2 Credit Valuation Adjustment (CVA)

🟡 **Definition:**

> **CVA** = the present value of the expected cost to the bank from a **counterparty default**:
>
> $$\text{CVA} \approx \sum_{i=1}^{n} (1-R)\,\lambda_i\,\Delta t_i\,E_i$$
>
> where:
> - $R$ = recovery rate
> - $\lambda_i$ = counterparty's hazard rate (default intensity) in period $i$
> - $E_i$ = expected exposure to the counterparty in period $i$
> - $\Delta t_i$ = length of period $i$
>
> **Expected exposure** $E_i$ = expected max(portfolio value, 0) at time $t_i$, after netting.

The **true derivative price** = clean price − CVA.

---

## 9.3 Debit Valuation Adjustment (DVA)

🟡 **Definition:**

> **DVA** = the present value of the expected gain to the bank from the possibility that **the bank itself defaults**:
>
> $$\text{DVA} \approx \sum_{i=1}^{n} (1-R_b)\,\lambda_b^i\,\Delta t_i\,E_b^i$$
>
> where $\lambda_b^i$ = bank's own hazard rate, $E_b^i$ = expected negative exposure (counterparty's exposure to the bank).
>
> DVA is controversial: a bank's creditworthiness deteriorating increases DVA (a "gain"), which seems counterintuitive. In practice: Price = Clean − CVA + DVA.

---

## 9.4 Collateral and Credit Support Annexes (CSAs)

🟡 **Definition:**

> A **CSA** (Credit Support Annex) specifies how collateral is calculated and what forms are acceptable. Key parameter: the **cure period (margin period of risk)** = number of days from when a counterparty stops posting collateral to when the master agreement is terminated and a settlement amount is computed (typically 10–20 days).
>
> Even with collateral, there is residual credit risk during the cure period.

---

## 9.5 Funding Valuation Adjustment (FVA) and Margin Valuation Adjustment (MVA)

🟡 **Definition:**

> **FVA** (Funding Valuation Adjustment) = adjustment for the **cost of funding** derivatives positions. When a bank enters a derivatives transaction that has positive value (an asset), it may need to post collateral to other counterparties or fund the margin requirement — this has a cost.
>
> **MVA** (Margin Valuation Adjustment) = adjustment for the cost of posting **initial margin** to central counterparties (CCPs). As regulatory requirements for initial margin have increased, MVA has become material.

Together, FVA + MVA capture the **total funding cost** of a derivatives portfolio.

---

## 9.6 Capital Valuation Adjustment (KVA)

🟡 **Definition:**

> **KVA** (Capital Valuation Adjustment) = charge to a derivative transaction for the **incremental regulatory capital** it requires. Banks must hold capital against their derivatives exposures (e.g., under Basel III/IV). This capital is costly (it must earn a return) → KVA represents the present value of the expected return required on the incremental capital over the life of the transaction.

---

## Summary

| XVA | What It Adjusts For | Sign (from bank's perspective) |
|-----|--------------------|-----------------------------|
| CVA | Counterparty default risk | Negative (cost) |
| DVA | Bank's own default risk | Positive (benefit) |
| FVA | Funding cost of unsecured derivatives | Negative (cost) |
| MVA | Cost of posting initial margin | Negative (cost) |
| KVA | Cost of regulatory capital | Negative (cost) |

**Full price** = Clean BSM price − CVA + DVA − FVA − MVA − KVA

---

## Total Value Adjustment

In practice, banks calculate the "XVA desk" P&L separately from the trading desk's "clean" P&L. The XVA desk manages the credit, funding, and capital adjustments across the entire derivatives book.
