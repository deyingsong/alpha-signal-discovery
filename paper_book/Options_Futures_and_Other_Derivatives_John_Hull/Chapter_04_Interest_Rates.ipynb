## Overview

This chapter covers the fundamental building blocks of fixed-income analysis: interest rate types, compounding conventions, yield curve construction, forward rates, duration, convexity, and term structure theories.

---

## 4.1 Types of Interest Rates

🟡 **Definition:**

> An **interest rate** defines the amount a borrower promises to pay the lender. For a given currency, many rates coexist: mortgage rates, deposit rates, prime rates, etc. The higher the **credit risk**, the higher the interest rate promised. The extra amount above the risk-free rate is the **credit spread**. Rates are often expressed in **basis points** (1 bp = 0.01%).

🟡 **Definition (Treasury Rates):**

> **Treasury rates** = rates earned on Treasury bills and bonds — instruments governments use to borrow in their own currency (e.g., U.S. T-bills, Japanese JGBs). Often treated as proxies for risk-free rates, though OIS rates have become the preferred risk-free benchmark.

🟡 **Definition (Overnight Reference Rates / OIS):**

> Banks maintain accounts at the central bank; those with surplus lend overnight to those with shortfalls. In the U.S., this **federal funds rate** is the overnight rate. **OIS (Overnight Index Swap)** rates derived from these overnight rates are now the standard risk-free benchmark, replacing LIBOR.

🟡 **Definition (Repo Rate):**

> In a **repurchase agreement (repo)**, a party sells securities and agrees to buy them back at a slightly higher price later. The implied interest rate is the **repo rate**. Because repos are collateralized, they are nearly risk-free. The most common type is an **overnight repo**; longer maturities are **term repos**.

🟡 **Definition (LIBOR):**

> **LIBOR** (London Interbank Offered Rate) = the rate at which major banks estimated they could borrow from each other. Was the key floating reference rate for decades but is being phased out (end of 2021) in favor of overnight rates (SOFR, SONIA, etc.).

🟢 **Comment:**

> The risk-free rate plays a central role in derivatives pricing — the standard approach sets up a riskless portfolio and argues its return must equal the risk-free rate. OIS rates (nearly risk-free, collateralized) are now the standard choice.

## 4.2 Measuring Interest Rates

🟡 **Property (Continuous Compounding):**

> As compounding frequency $m \to \infty$, an amount $A$ invested at rate $R$ for $n$ years grows to:
>
> $$A\,e^{Rn}$$
>
> Continuous compounding is mathematically convenient and used throughout derivatives theory.

**Conversion between compounding frequencies:**

$$R_c = m\ln\!\left(1 + \frac{R_m}{m}\right), \quad R_m = m\left(e^{R_c/m} - 1\right)$$

where $R_c$ = continuously compounded rate, $R_m$ = rate with compounding $m$ times per year.

---

## 4.3 Zero Rates

🟡 **Definition:**

> The **$n$-year zero-coupon (zero) rate** = rate of interest earned on an investment starting today and maturing in exactly $n$ years, with all interest and principal paid at maturity (no intermediate payments). Also called the **$n$-year spot rate**.

Zero rates form the foundation for pricing all fixed-income instruments.

---

## 4.4 Bond Pricing

Given zero rates $r_1, r_2, \ldots, r_n$, a coupon bond price is:

$$B = \sum_{i=1}^{n} c_i\,e^{-r_i t_i}$$

where $c_i$ = cash flow at time $t_i$.

🟡 **Definition (Bond Yield):**

> The **bond yield** $y$ = single discount rate that, when applied to all cash flows, gives the observed market price:
>
> $$B = \sum_{i=1}^{n} c_i\,e^{-y t_i}$$

🟡 **Definition (Par Yield):**

> The **par yield** = coupon rate that makes the bond price equal its face value (par). For a bond with semiannual coupons and face value $100$:
>
> $$100 = \frac{c}{2}\sum_{i=1}^{2n} e^{-r_i t_i} + 100\,e^{-r_{2n}\cdot n}$$
>
> Solving for $c$ gives the par yield.

---

## 4.5 Determining Zero Rates (Bootstrap Method)

🟡 **Property:**

> The **bootstrap method** extracts zero rates from bond prices sequentially. Start with the shortest maturity bond (often a T-bill) to get the short zero rate, then work outward:
>
> $$r_n = -\frac{1}{t_n}\ln\!\left(\frac{B - \sum_{i<n} c_i e^{-r_i t_i}}{c_n}\right)$$

This builds the **zero curve (spot curve)** from market bond prices.

---

## 4.6 Forward Rates

🟡 **Definition:**

> **Forward rates** = interest rates implied by current zero rates for future periods. For the period from $T_1$ to $T_2$:
>
> $$R_F = \frac{R_2 T_2 - R_1 T_1}{T_2 - T_1}$$
>
> where $R_1$, $R_2$ are zero rates for maturities $T_1$, $T_2$ (continuously compounded).

🟢 **Comment:**

> If an investor believes future rates will differ from today's forward rates, arbitrage opportunities arise. One strategy involves entering a **Forward Rate Agreement (FRA)**: locking in the forward rate today for borrowing/lending in the future. If actual rates differ from the forward rate, a profit or loss results.

---

## 4.7 Forward Rate Agreements (FRAs)

🟡 **Definition:**

> An **FRA** is an agreement to exchange a predetermined fixed rate $R_K$ for the reference rate $R_M$ observed at time $T_1$, applied to a notional principal $L$ over a period $[T_1, T_2]$:
>
> - FRA payoff (to fixed receiver) at $T_2$: $L(R_K - R_M)(T_2 - T_1)$

🟢 **Property:**

> An FRA can be valued by assuming the forward rate $R_F$ will be realized:
>
> - Value (to fixed receiver) = $L(R_K - R_F)(T_2 - T_1) e^{-r_2 T_2}$
>
> This is the **FRA valuation principle** — forward rates are used as certainty-equivalent expected future rates under the risk-neutral measure.

---

## 4.8 Duration

🟡 **Definition:**

> The **duration** $D$ of a bond measures the weighted average time to receipt of cash flows (weights = PV of cash flow / bond price):
>
> $$D = \frac{\sum_{i=1}^{n} t_i\, c_i\, e^{-y t_i}}{B}$$
>
> For a zero-coupon bond: $D = T$ (maturity).

🟡 **Definition (Modified Duration):**

> $$D^* = \frac{D}{1 + y/m}$$
>
> where $m$ = compounding frequency. Gives the price sensitivity:
>
> $$\Delta B \approx -B\, D^*\, \Delta y \quad (4.13)$$

🟢 **Comment:**

> When duration is used for bond **portfolios**, there is an implicit assumption that all yields change by the same amount (parallel shift). This holds well for small changes but breaks down for large or non-parallel moves.

---

## 4.9 Convexity

🟡 **Property:**

> The bond price–yield relationship is **convex** (curved). Duration gives a linear approximation; **convexity** captures the curvature:
>
> $$C = \frac{1}{B}\sum_{i=1}^{n} c_i\, t_i^2\, e^{-y t_i}$$
>
> Improved price change approximation:
>
> $$\Delta B \approx -B\,D^*\,\Delta y + \frac{1}{2}B\,C\,(\Delta y)^2$$
>
> Higher convexity is always beneficial (for the same duration, a bond with higher convexity gains more when yields fall and loses less when yields rise).

---

## 4.10 Theories of the Term Structure

🟡 **Property (Expectations Theory):**

> Long-term rates reflect expected future short-term rates. The forward rate for a future period equals the **expected future zero rate** for that period.

🟡 **Property (Market Segmentation Theory):**

> Short-, medium-, and long-term bond markets are segmented by investor preferences. Supply and demand in each segment independently determines rates — no relationship across maturities.

🟡 **Property (Liquidity Preference Theory):**

> Investors prefer **short-term** instruments (more liquid); borrowers prefer **long-term** funding. To attract long-term investors, borrowers must offer a **liquidity premium** above expected future short rates. This causes the yield curve to be upward-sloping on average.

🟢 **Comment:**

> Liquidity preference theory explains the typical upward-sloping yield curve. It also explains why term structure management at banks is critical: banks borrow short-term (deposits) and lend long-term (mortgages). A sudden rise in rates can devastate net interest income.
