## Overview

Equilibrium models derive the interest rate process from economic assumptions. They typically don't fit the current term structure exactly (unlike no-arbitrage models in Chapter 32) but provide theoretical insight into yield curve dynamics.

---

## 31.1 Background

🟢 **Comment:**

> **Equilibrium models** start with assumptions about economic variables (e.g., preferences, technology) and derive the risk-neutral process for the short rate $r$. Key characteristics:
> - Model parameters are determined by theory/estimation, not market fitting
> - Do **not** automatically fit the current term structure
> - Used primarily for understanding yield curve dynamics and relative value

🟡 **Definition:**

> The **risk-free short rate** $r(t)$ = instantaneous rate applying to an infinitesimally short time period at time $t$. All bond prices and derivative prices depend on the process followed by $r$.

### Bond Pricing Under Short Rate Models

🟢 **Comment:**

> For equilibrium models with affine structure, zero-coupon bond prices take the form:
>
> $$P(t,T) = A(t,T)\,e^{-B(t,T)r(t)}$$
>
> → The zero rate: $R(t,T) = -\frac{\ln A + B\cdot r}{T-t}$

---

## 31.2 One-Factor Models

### Rendleman–Bartter Model

🟡 **Definition:**

> $$dr = \mu r\,dt + \sigma r\,dz$$
>
> (Geometric Brownian motion for $r$.) Simple, but $r$ cannot become negative. However, it also cannot exhibit **mean reversion** — interest rates can drift arbitrarily far, which is unrealistic.

### Vasicek Model

🟡 **Definition:**

> $$dr = a(b - r)\,dt + \sigma\,dz$$
>
> - $a$ = speed of mean reversion
> - $b$ = long-run mean of $r$
> - $\sigma$ = volatility
>
> The short rate is pulled toward $b$ at speed $a$. This is the **Ornstein–Uhlenbeck process**. Allows negative rates (a weakness), but has desirable mean-reversion. Yields analytical formulas for bond prices.

**Bond price:**

$$P(t,T) = A(t,T)e^{-B(t,T)r(t)}$$

$$B(t,T) = \frac{1-e^{-a(T-t)}}{a}, \quad \ln A(t,T) = \left(B - T + t\right)\left(ab - \frac{\sigma^2}{2a^2}\right) - \frac{\sigma^2 B^2}{4a}$$

### Cox–Ingersoll–Ross (CIR) Model

🟡 **Definition:**

> $$dr = a(b - r)\,dt + \sigma\sqrt{r}\,dz$$
>
> Same mean-reverting drift as Vasicek, but the diffusion term $\sigma\sqrt{r}$ ensures:
> - Volatility scales with $\sqrt{r}$ (higher volatility when rates are high)
> - **$r$ cannot become negative** (when $2ab > \sigma^2$)
>
> This is more realistic than Vasicek for interest rates.

### Two-Factor Model (Extended Vasicek)

🟡 **Definition:**

> $$dr = (\theta - ar)\,dt + \sigma_1\,dz_1$$
> $$d\theta = -b\theta\,dt + \sigma_2\,dz_2$$
>
> The **reversion level** $\theta/a$ is itself a stochastic process, making the model two-factor. This allows for a wider range of yield curve shapes and correlation structures.

---

## 31.3 Real-World vs. Risk-Neutral Processes

In equilibrium models, the real-world drift of $r$ is:

$$\mu_r = a(b - r) - \lambda\sigma$$

where $\lambda$ is the market price of interest rate risk. The risk-neutral drift sets $\lambda = 0$:

$$\tilde{\mu}_r = a(b - r)$$

This separation allows calibration to cross-sectional bond prices (risk-neutral) and time-series data (real-world).

---

## 31.4 Estimating Parameters

Parameters ($a, b, \sigma$) can be estimated from:
1. **Time-series** of short rates (maximum likelihood)
2. **Cross-section** of bond prices (fitting the yield curve)

The two approaches typically give different estimates, reflecting the difference between real-world and risk-neutral processes.

---

## 31.5 More Sophisticated Models

Multi-factor models (e.g., two-factor CIR, Longstaff–Schwartz) provide more flexibility in fitting yield curve shapes and correlations. However, they are harder to calibrate and less tractable analytically.

## Summary

| Model | Process | Mean Reversion | $r \geq 0$? | Bond Price |
|-------|---------|----------------|-------------|-----------|
| Rendleman–Bartter | $dr = \mu r\,dt + \sigma r\,dz$ | No | Yes | No closed form |
| Vasicek | $dr = a(b-r)\,dt + \sigma\,dz$ | Yes | No (can go negative) | Analytical |
| CIR | $dr = a(b-r)\,dt + \sigma\sqrt{r}\,dz$ | Yes | Yes (if $2ab > \sigma^2$) | Analytical |
| Two-factor Vasicek | $dr, d\theta$ | Yes (stochastic level) | No | Analytical |
