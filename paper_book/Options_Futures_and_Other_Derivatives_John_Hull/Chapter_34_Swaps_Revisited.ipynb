# Chapter 34: Swaps Revisited

> **Color Convention**  
> 🟡 **Yellow** = Definition / Theorem / Property  
> 🟢 **Green** = Comment / Application / Insight

---

## Overview

This chapter surveys variations on the standard (vanilla) interest rate swap, including swaps with variable notional, compounding swaps, currency swaps, equity swaps, swaps with embedded options, and other exotic swap structures.

---

## 34.1 Variations on the Vanilla Deal

🟡 **Definition:**

> **Step-up swaps**: notional principal increases over time (e.g., to match a loan drawdown schedule). **Step-down (amortizing) swaps**: notional principal decreases over time (e.g., to match a loan amortization).

🟡 **Definition:**

> **Basis swaps**: exchange interest at one floating reference rate for interest at another floating rate (e.g., 3M SOFR vs. 6M SOFR). Useful for entities whose assets and liabilities reference different floating benchmarks.

---

## 34.2 Compounding Swaps

🟡 **Definition:**

> A **compounding swap**: instead of paying out interest at each reset date, payments are reinvested (compounded) at a specified rate until the swap's final maturity date. Only one net payment is made at the end.
>
> Valuation: use adjusted forward rates incorporating the compounding, derived from equation (30.3) for timing/convexity adjustments.

---

## 34.3 Currency and Nonstandard Swaps

🟡 **Definition:**

> A **differential (diff) swap** or cross-currency swap: an interest rate in one currency (e.g., USD SOFR) applied to a principal in a different currency (e.g., EUR). This involves a **quanto adjustment** as discussed in Chapter 30 — the drift of the foreign rate must be adjusted for the correlation between the foreign rate and the exchange rate.

---

## 34.4 Equity Swaps

🟡 **Definition:**

> An **equity swap**: one party pays the total return on an equity index (or basket of stocks) on a notional principal, while the other pays fixed or floating interest.
>
> Applications:
> - Fund managers use equity swaps to **increase or decrease index exposure** without buying/selling the physical index
> - Allows cross-border equity exposure with favorable tax/regulatory treatment
>
> Valuation: equity leg value = notional × (current index level / initial index level); at each reset date the notional is effectively "reset" to the current index value.

---

## 34.5 Swaps with Embedded Options

🟡 **Definition:**

> **Index amortizing rate swap (IAS)** (also called indexed principal swap): the notional principal decreases according to a schedule linked to the **level of interest rates**. The lower the rate, the greater the principal reduction (mimicking mortgage prepayments). Investors effectively sell a complex option to the fixed-rate payer.

Other swaps with embedded options:
- **Accrual swaps**: interest accrues only on days when a reference rate stays within a specified range
- **Cancelable swaps**: one party has the right to cancel the swap at one or more future dates (embedded Bermudan swaption)
- **Extendable swaps**: one party can extend the maturity of the swap (another embedded option)

---

## 34.6 Other Swaps

🟡 **Definition (Commodity Swap):**

> A **commodity swap**: one party pays a fixed price for a commodity (e.g., $70/barrel for 100,000 barrels of oil per quarter) while the other pays the market price. Used by producers and consumers to hedge commodity price risk.

Other notable structures:
- **Volatility swaps**: exchange realized volatility for a fixed volatility strike
- **Variance swaps**: exchange realized variance for a fixed variance rate (discussed in Chapter 26)

---

## Summary

| Swap Type | Key Feature | Valuation Approach |
|-----------|-------------|-------------------|
| Step-up / amortizing | Variable notional | Standard, adjusted for notional schedule |
| Basis swap | Two floating rates exchanged | Bootstrapping from FRA/futures markets |
| Compounding swap | Interest compounds to maturity | Adjusted forward rates (timing adjustment) |
| Diff swap | Foreign rate on domestic principal | Quanto adjustment |
| Equity swap | Pay index return, receive fixed/floating | Reset equity leg to index value |
| Index amortizing | Principal linked to rate level | Monte Carlo / tree (path-dependent) |
| Cancelable swap | Embedded Bermudan swaption | Bermudan swaption valuation |
| Commodity swap | Fixed vs. floating commodity price | Futures curve pricing |
