## Overview

This chapter presents two important frameworks for modeling the entire forward rate curve: the Heath–Jarrow–Morton (HJM) model and the BGM (LIBOR Market Model). It also covers agency mortgage-backed securities.

---

## 33.1 The Heath–Jarrow–Morton (HJM) Model

### Framework

🟡 **Property (Key HJM Result):**

> Consider the risk-neutral process for the instantaneous forward rate $F(t, T)$ for maturity $T$:
>
> $$dF(t,T) = \mu(t,T)\,dt + \sigma(t,T)\,dz$$
>
> For a zero-coupon bond $P(t,T)$ to be correctly priced (no arbitrage), **the drift is fully determined by the volatility**:
>
> $$\mu(t,T) = \sigma(t,T)\int_t^T \sigma(t,\tau)\,d\tau$$
>
> This is the central HJM result: once the volatility function $\sigma(t,T)$ is specified, the entire forward rate dynamics are determined under the risk-neutral measure.

🟢 **Comment:**

> HJM has two drawbacks:
> 1. It is expressed in terms of **instantaneous forward rates** — not directly observable
> 2. It is **difficult to calibrate** to market instruments (caps, swaptions)
>
> These drawbacks motivated the development of the LIBOR Market Model (BGM).

## 33.2 The BGM Model (LIBOR Market Model)

🟢 **Comment:**

> **BGM** (Brace, Gatarek, Musiela — also called the LIBOR Market Model) overcomes HJM's limitations by modeling **discrete** forward LIBOR rates $F_k$ directly — the rates that correspond to observable market instruments.

### LIBOR Market Model Setup

Forward LIBOR rate $F_k$ = forward rate for period $[t_k, t_{k+1}]$, observable from Eurodollar futures / FRA markets.

Under the numeraire of the bond maturing at $t_{k+1}$ (**rolling risk-neutral world**), each $F_k$ is lognormal:

$$\frac{dF_k}{F_k} = \sigma_k\,dz_k$$

🟡 **Property (Drift in Rolling Risk-Neutral World):**

> When pricing using the money-market account as numeraire (rolling risk-neutral world), the drift of $F_k$ is:
>
> $$\frac{dF_k}{F_k} = \left(-\sum_{j=k+1}^{n} \frac{\delta_j F_j \rho_{k,j} \sigma_k \sigma_j}{1 + \delta_j F_j}\right)\,dt + \sigma_k\,dz_k$$
>
> where $\delta_j = t_{j+1} - t_j$ and $\rho_{k,j}$ is the correlation between $F_k$ and $F_j$.

This drift term allows the model to be **simulated via Monte Carlo** — jump from one reset date to the next, applying the drift correction at each step.

### Caps Pricing

Since each $F_k$ is lognormal under its natural measure $P(0, t_{k+1})$, the BGM model gives exactly **Black's caplet formula** — so it is automatically consistent with the caps market.

### Swaptions Pricing

Swaptions are priced approximately using the **swap rate approximation**: the forward swap rate is approximately lognormal as a weighted average of the forward LIBOR rates.

🟢 **Comment:**

> The BGM approximation for the swap rate volatility:
>
> $$\sigma_s \approx \frac{\sum_k w_k \sigma_k}{\text{swap annuity}}$$
>
> This allows calibration to both caps and swaptions simultaneously.

### Non-Standard Caps

🟢 **Application:**

> BGM can value non-standard caps:
> - **Ratchet cap**: Cap rate for each caplet = previous reset rate + spread
> - **Sticky cap**: Cap rate = min(previous reset rate + spread, previous cap rate)
> - **Flexi cap**: Limit on total number of caplets that can be exercised
>
> These require Monte Carlo simulation since payoffs depend on the history of rates.

🟢 **Application:**

> The Monte Carlo implementation of BGM benefits from variance reduction via antithetic variates and control variates (using vanilla cap/swaption prices as control variates).

## 33.3 Agency Mortgage-Backed Securities

### MBS Structure

🟡 **Definition:**

> Agency MBS are pools of residential mortgages. Cash flows depend on homeowner **prepayments**. Prepayments occur when:
> - Mortgage rates fall → homeowner refinances at lower rate
> - House is sold
>
> The **prepayment function** describes the prepayment rate as a function of the interest rate environment. This makes MBS path-dependent.

### CMOs

🟡 **Definition:**

> A **Collateralized Mortgage Obligation (CMO)** divides MBS cash flows into tranches with different prepayment risk exposures. Tranches bear different amounts of prepayment risk:
> - Senior tranches: less prepayment risk (more stable cash flows)
> - Subordinate tranches: more prepayment risk

### Option-Adjusted Spread (OAS)

🟡 **Property:**

> The **OAS** measures the spread over the risk-free curve that equates the model price of an MBS or callable bond to its market price, after adjusting for embedded options:
>
> $$\text{Market price} = E\left[\sum_t \frac{\text{Cash flow}_t}{(1+r_t+\text{OAS})^t}\right]$$
>
> OAS strips out the value of embedded options; it represents the pure credit/liquidity spread.

## Summary

| Model | Key Feature | Main Use |
|-------|-------------|----------|
| HJM | Drift determined by volatility; models instantaneous forward rates | Theoretical framework |
| BGM / LMM | Models discrete LIBOR rates; calibrates to caps | Practical cap/swaption pricing |
| Rolling risk-neutral (BGM) | Simulation jumps between reset dates | Monte Carlo for path-dependent rates |
| CMO | Tranched prepayment risk | MBS structuring |
| OAS | Spread after option adjustment | MBS relative value |
