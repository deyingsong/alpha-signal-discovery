## Overview

This chapter surveys how options can be combined with each other and with the underlying stock to create a wide variety of payoff profiles. Key strategies: covered calls, principal-protected notes, spreads (bull, bear, butterfly, calendar, diagonal), and combinations (straddles, strangles, strips, straps).

---

## 12.1 Principal-Protected Notes

🟡 **Definition:**

> A **principal-protected note** = structured product guaranteeing return of initial investment (the "principal") regardless of performance, while offering upside linked to a risky asset.
>
> Construction: invest $K e^{-rT}$ in risk-free zero-coupon bond (grows to $K$ at maturity) + use the remaining cash to buy a call option on the risky asset.

🟡 **Property:**

> A **simple principal-protected note** can be created with:
> - Bond position covering the principal (guaranteed return of $K$)
> - Call option providing upside participation

## 12.2 Strategies Involving a Single Option and Stock

🟡 **Definition (Covered Call):**

> A **covered call** = long stock + short (written) call.
>
> The long stock "covers" the potential obligation on the short call.
>
> Profit profile:
> - If $S_T > K$: profit capped at $K + c - S_0$ (stock gains offset by call loss)
> - If $S_T \leq K$: profit = $S_T - S_0 + c$ (no call exercise; retain the premium)
>
> Motivation: generate income on a stock position you are willing to sell at $K$.

## 12.3 Spreads

### Bull Spread

🟡 **Definition:**

> A **bull spread** = buy call at $K_1$ + sell call at $K_2$ ($K_2 > K_1$), same expiration.
>
> | $S_T$ range | Payoff |
> |------------|--------|
> | $S_T < K_1$ | 0 |
> | $K_1 \leq S_T \leq K_2$ | $S_T - K_1$ |
> | $S_T > K_2$ | $K_2 - K_1$ |
>
> Net cost: $c_1 - c_2 > 0$ (limited cost). Maximum gain = $K_2 - K_1 - (c_1 - c_2)$.
>
> **Motivation**: bullish view; lower cost than buying the call outright (premium from short call offsets cost of long call). Can also be created with puts.

### Bear Spread

🟡 **Definition:**

> A **bear spread** = buy put at $K_2$ + sell put at $K_1$ ($K_2 > K_1$), same expiration.
>
> Profits if stock falls; loss limited to net premium paid. Can also be created with calls (short call at $K_1$, long call at $K_2$ → negative initial cash flow for call bear spread).

### Butterfly Spread

🟡 **Definition:**

> A **butterfly spread** uses three strikes $K_1 < K_2 < K_3$ (usually $K_2 = (K_1+K_3)/2$):
>
> Long call at $K_1$, short **2** calls at $K_2$, long call at $K_3$.
>
> Payoff peaks at $K_2$ (maximum if $S_T = K_2$), zero at $K_1$ and $K_3$.
>
> **Motivation**: profits if stock stays near current price; appropriate for low-volatility view.

### Calendar Spread

🟡 **Definition:**

> A **calendar spread** = sell short-maturity call + buy long-maturity call, same strike.
>
> Profits if stock price stays near the strike (the short call decays faster than the long call). Three variants:
> - **Neutral calendar spread**: strike near current price
> - **Bullish calendar spread**: higher strike
> - **Bearish calendar spread**: lower strike

🟡 **Property:**

> A **reverse calendar spread**: buy short-maturity option + sell long-maturity option. Profits if stock price moves significantly away from the strike at expiry of the short option.

### Diagonal Spread

🟡 **Definition:**

> A **diagonal spread** uses two calls (or puts) with **different strikes AND different maturities**. More flexibility in profit profiles than a calendar spread (same strike) or a vertical spread (same maturity).

## 12.4 Combinations

### Straddle

🟡 **Definition:**

> A **straddle** = long call + long put, same strike $K$, same expiration.
>
> Profits if $S_T$ moves significantly away from $K$ in either direction. Loss (premium paid) if $S_T \approx K$.
>
> - **Bottom straddle (straddle purchase)**: long call + long put → bet on high volatility
> - **Top straddle (straddle write)**: short call + short put → bet on low volatility (highly risky)

🟡 **Property:**

> A straddle is most profitable when the market expects a large move (e.g., earnings announcement) but the actual move is even larger. A trader who buys a straddle before a known event is betting that the implied volatility understates the realized move.

### Strips and Straps

🟡 **Definition:**

> - **Strip** = long 1 call + long 2 puts (same $K$, same $T$): bets on a large move but more bearish (downward move pays more)
> - **Strap** = long 2 calls + long 1 put (same $K$, same $T$): bets on a large move but more bullish (upward move pays more)

### Strangle

🟡 **Definition:**

> A **strangle** (bottom vertical combination) = long call at $K_2$ + long put at $K_1$, with $K_2 > K_1$, same expiration.
>
> Cheaper than a straddle (both options are out-of-the-money) but requires a larger move to profit.
>
> | $S_T$ range | Payoff |
> |------------|--------|
> | $S_T < K_1$ | $K_1 - S_T$ |
> | $K_1 \leq S_T \leq K_2$ | 0 (loss of premium) |
> | $S_T > K_2$ | $S_T - K_2$ |
>
> **Top strangle** (sell strangle) = short call at $K_2$ + short put at $K_1$. Profits if the stock stays between $K_1$ and $K_2$; appropriate for low-volatility view; cheaper premium received than a short straddle.

## Summary

| Strategy | Position | View |
|----------|----------|------|
| Covered call | Long stock + short call | Neutral to mildly bullish; income |
| Bull spread (calls) | Long call $K_1$, short call $K_2$ | Bullish (limited cost/gain) |
| Bear spread (puts) | Long put $K_2$, short put $K_1$ | Bearish (limited cost/gain) |
| Butterfly spread | Long $K_1$, short 2$K_2$, long $K_3$ | Neutral; low volatility |
| Calendar spread | Short near, long far (same strike) | Neutral; time decay |
| Straddle | Long call + long put (same $K$) | High volatility |
| Strangle | Long OTM call + long OTM put | High volatility (cheaper than straddle) |
| Strip | 1 call + 2 puts | High vol, bearish bias |
| Strap | 2 calls + 1 put | High vol, bullish bias |
