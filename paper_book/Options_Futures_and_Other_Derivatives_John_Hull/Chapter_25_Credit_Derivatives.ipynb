## Overview

Credit derivatives allow the trading of credit risk separately from market risk. Market grew from $800B (2000) to $50T (2007 peak), settling at ~$7.5T by 2019. Key products: CDS, credit indices, basket CDS, total return swaps, and CDOs.

---

## 25.1 Credit Default Swaps (CDS)

🟡 **Definition:**

> A **CDS** provides insurance against default of a **reference entity**:
> - **Protection buyer** pays periodic premiums (the CDS spread) until default or maturity
> - **Protection seller** pays $(1-R) \times \text{notional}$ upon a credit event
>
> Credit events include: bankruptcy, failure to pay, restructuring.
>
> The **CDS spread** (quoted in bps/year) is set so the contract has zero initial value.

🟡 **Property (CDS–Bond Basis):**

> $$\text{CDS–bond basis} = \text{CDS spread} - \text{bond yield spread}$$
>
> In theory, the basis should be zero (arbitrage: long bond + buy CDS $\approx$ risk-free). In practice, the basis deviates, especially during crises (funding constraints, counterparty risk, cheapest-to-deliver option).

---

## 25.2 Valuation of Credit Default Swaps

The fair CDS spread $\bar{s}$ equates PV(premium leg) = PV(protection leg):

**Premium leg** (assuming payments at times $t_j$, spacing $\Delta t$):

$$\text{PV}_{\text{prem}} = \bar{s}\sum_j e^{-rt_j}\Delta t_j [1-q(t_j)] + \bar{s}\sum_j e^{-rt_j}\frac{\Delta t_j}{2}[q(t_j)-q(t_{j-1})]$$

(Second term captures accrued premium paid at default.)

**Protection leg**:

$$\text{PV}_{\text{prot}} = (1-R)\sum_j e^{-rt_j}[q(t_j)-q(t_{j-1})]$$

Setting equal and solving for $\bar{s}$ gives the fair CDS spread.

## 25.3 Credit Indices

🟢 **Application:**

> Standard index portfolios allow portfolio-level credit trading:
> - **CDX NA IG**: 125 investment-grade North American companies
> - **iTraxx Europe**: 125 investment-grade European companies
>
> The index spread approximates the equal-weighted average individual CDS spread. Indices allow efficient entry/exit from credit exposure without trading 125 individual CDS contracts.

---

## 25.4 Fixed Coupons in CDS

🟢 **Application:**

> In practice, CDS trade with **standard coupons** (100 bps or 500 bps). The difference between the fair spread and the coupon is settled upfront:
> 1. Assume 4 quarterly payments in arrears at the standard coupon
> 2. Compute PV(payments at standard coupon) and PV(protection)
> 3. Upfront payment = difference, adjusted to make the contract fair

---

## 25.5 CDS Forwards and Options

- **CDS forward**: commitment to enter a CDS at a specified spread at a future date
- **CDS option**: right (not obligation) to enter a CDS at a specified spread; knockout on default

---

## 25.6 Basket Credit Default Swaps

🟡 **Definition:**

> - **Add-up basket CDS**: pays off when *any* reference entity defaults
> - **First-to-default (FtD) CDS**: pays off only when the *first* default occurs
> - **$k$th-to-default CDS**: pays off when the $k$th default occurs
>
> Basket CDS values are highly sensitive to default correlation. Higher correlation → FtD CDS worth less (less likely any single default triggers it in isolation when defaults cluster).

---

## 25.7 Total Return Swaps

🟡 **Definition:**

> A **total return swap (TRS)** exchanges the total return on a reference asset (coupons + price changes) for a floating rate (e.g., SOFR) plus a spread.
>
> - TRS seller (protection buyer) receives floating + spread; pays total return
> - TRS buyer (protection seller) receives total return; pays floating + spread
>
> This allows off-balance-sheet leveraged exposure to the reference asset.

---

## 25.8 Collateralized Debt Obligations (CDOs)

🟡 **Definition:**

> A **CDO** creates tranched exposure from a portfolio of bonds (cash CDO) or CDS (synthetic CDO):
>
> - **Senior tranche**: absorbs losses *last*; lowest risk and yield; typically AAA rated
> - **Mezzanine tranche**: intermediate risk and yield
> - **Equity tranche**: absorbs *first losses*; highest risk; unrated; acts as first-loss piece

🟡 **Definition (Synthetic CDO):**

> Uses CDS contracts rather than physical bonds. An **imaginary reference portfolio** is defined; tranches pay/receive based on simulated losses to this portfolio.
>
> **Attachment point** $a_L$ and **detachment point** $a_H$ define each tranche: absorbs losses between $a_L\%$ and $a_H\%$ of the total portfolio notional.

---

## 25.9 Role of Correlation

🟢 **Comment:**

> Default correlation is critical for CDO tranche pricing:
> - **Low correlation**: losses are spread across names → junior tranche very risky (high expected loss), senior tranche very safe
> - **High correlation**: all-or-nothing outcomes → junior tranche less risky (often survives), senior tranche riskier (occasionally absorbs all losses)
>
> The equity tranche is long correlation; the senior tranche is short correlation.

---

## 25.10 Valuation of a Synthetic CDO

**Standard market model (Gaussian copula):**

1. Each company $i$'s normalized time to default: $x_i = \sqrt{\rho}\,F + \sqrt{1-\rho}\,Z_i$

2. Condition on common factor $F$: conditional on $F$, defaults are independent

3. Conditional default probability for company $i$ given $F$:

$$p_i(F) = N\!\left(\frac{N^{-1}(Q_i(t)) - \sqrt{\rho}\,F}{\sqrt{1-\rho}}\right)$$

4. For each value of $F$, compute expected tranche loss analytically (binomial / normal approximation)

5. Integrate over $F$ (standard normal) to get unconditional expected tranche loss

🟢 **Application (Base Correlation):**

> **Base correlation**: the correlation $\rho$ that makes a $[0, a_H]$ tranche value match the market quote. Allows interpolation across the capital structure and comparison across CDOs.

🟢 **Application ($k$th-to-Default CDS):**

> Condition on $F$; compute the conditional probability that exactly $k-1$ of $n$ companies have defaulted by time $t_j$; price the $k$th-to-default payoff accordingly.

---

## 25.11 Alternatives to the Standard Market Model

🟡 **Property:**

> - **Structural models**: Correlated asset values drive defaults (Merton-type); default intensities can jump
> - **Reduced form models**: Correlated stochastic hazard rates with mean-reversion and jumps; require simulation and calibration
> - **Top-down models**: Model the portfolio loss process directly without specifying individual names; more tractable for pricing tranches

## Summary

| Product | Description |
|---------|-------------|
| CDS | Insurance on a single reference entity |
| Basket / FtD CDS | Insurance triggered by $k$th default in a portfolio |
| Total return swap | Exchange total asset return for floating rate |
| CDO | Tranched exposure to bond/CDS portfolio |
