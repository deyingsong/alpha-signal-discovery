## Overview

This chapter covers interest rate futures, including Treasury bond futures and SOFR/Eurodollar futures, with applications to duration-based hedging and portfolio immunization.

---

## 6.1 Day Count and Quotation Conventions

🟡 **Definition:**

> A **day count convention** specifies how interest accrues between two dates. The convention is expressed as (numerator / denominator):
>
> | Convention | Used For |
> |------------|----------|
> | Actual/Actual (in period) | U.S. Treasury bonds |
> | Actual/360 | U.S. money market instruments (T-bills, LIBOR) |
> | 30/360 | U.S. corporate and municipal bonds |

🟢 **Comment (Actual/Actual):**

> For U.S. Treasury bonds, interest earned between two dates = actual days elapsed / actual days in coupon period × coupon amount. This can produce **surprising** results — some coupon periods have more days than others.

🟢 **Comment (Actual/360):**

> For money market instruments, the reference period is 360 days. **Surprising consequence**: a 365-day T-bill earns more interest than its stated rate implies because 365 > 360.

🟡 **Definition (Discount Rate Quotation):**

> Money market instruments are sometimes quoted as a **discount rate** — interest expressed as a percentage of **face value** (not as a percentage of investment amount). This understates the true yield relative to conventional bond quoting.
>
> $$\text{Discount rate} = \frac{\text{Face} - \text{Price}}{\text{Face}} \times \frac{360}{n}$$

## 6.2 Treasury Bond Futures

🟡 **Property (Clean vs. Dirty Price):**

> The **quoted (clean) price** excludes accrued interest. The **cash (dirty) price** = clean price + accrued interest.
>
> $$\text{Cash price} = \text{Quoted price} + \text{Accrued interest since last coupon}$$

🟡 **Definition (Deliverable Bonds):**

> Treasury bond futures allow delivery of any Treasury bond with **maturity between 15 and 25 years** (for 30-year futures). Treasury notes (1–10 year maturity) underlie T-note futures. The most actively traded are the 10-year, 5-year, and 2-year contracts.

🟡 **Definition (Conversion Factor):**

> The **conversion factor** (CF) for a deliverable bond standardizes prices across bonds with different coupons and maturities. When a bond is delivered, the short party receives:
>
> $$\text{Cash received} = Q_F \times CF + \text{Accrued interest}$$
>
> where $Q_F$ = quoted futures price. The CF is approximately the price of the bond per $1 of face value if it had a 6% yield (set by the exchange).

**Cheapest-to-Deliver (CTD)**: The short party delivers whichever eligible bond maximizes:

$$\text{Quoted bond price} - (Q_F \times CF)$$

## 6.3 Eurodollar and SOFR Futures

🟢 **Comment:**

> **Eurodollar futures** (based on 3-month LIBOR) have been the dominant short-term interest rate futures contract. As LIBOR is phased out (end of 2021), **SOFR futures** are replacing them. Trading volume in SOFR futures has been growing while Eurodollar volume declines.

SOFR futures are based on the **Secured Overnight Financing Rate** (the U.S. overnight repo rate) — essentially a 3-month compound of daily SOFR rates.

## 6.4 Duration-Based Hedging with Interest Rate Futures

🟡 **Property:**

> To hedge a bond portfolio against yield changes, the number of futures contracts needed is:
>
> $$N^* = \frac{P\,D_P}{V_F\,D_F} \quad (6.3)$$
>
> where:
> - $P$ = value of bond portfolio
> - $D_P$ = duration of portfolio at the hedge horizon
> - $V_F$ = price of one futures contract (notional)
> - $D_F$ = duration of the cheapest-to-deliver bond at futures maturity

This is the **duration-based hedge ratio** — it equalizes the dollar duration of the portfolio and the futures position.

**Derivation**: Under parallel yield shifts $\Delta y$:

$$\Delta P \approx -P\,D_P\,\Delta y, \quad \Delta V_F \approx -V_F\,D_F\,\Delta y$$

Setting $N^* \Delta V_F + \Delta P = 0$ gives $N^* = P D_P / (V_F D_F)$.

---

## 6.5 Duration Matching (Immunization)

🟡 **Property:**

> **Duration matching (immunization)**: a financial institution hedges interest rate risk by ensuring the duration of its **assets** equals the duration of its **liabilities**. This protects net worth against small parallel yield shifts.
>
> $$D_{\text{assets}} = D_{\text{liabilities}}$$
>
> Used by pension funds, insurance companies, and banks to manage their balance sheet interest rate exposure.
