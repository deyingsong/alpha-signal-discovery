# Chapter 20: Volatility Smiles and Volatility Surfaces

> **Color Convention**  
> 🟡 **Yellow** = Definition / Theorem / Property  
> 🟢 **Green** = Comment / Application / Insight

---

## Overview

This chapter examines how traders use the Black–Scholes–Merton (BSM) model in practice. Rather than a constant volatility, traders allow implied volatility to depend on **strike price** and **time to maturity**.

- **Volatility smile**: plot of implied vol vs. strike price for a fixed maturity
- **Volatility surface**: 3D plot of implied vol vs. both strike price and maturity

---

## 20.1 Implied Volatilities of Calls and Puts

Due to **put–call parity** (model-free), the implied vol of a European call equals that of a European put with the same strike and maturity:

$$p + S_0 e^{-qT} = c + K e^{-rT}$$

This means a volatility smile applies equally to calls and puts — no need to distinguish.

---

## 20.2 Volatility Smile for Foreign Currency Options

The FX volatility smile is roughly **U-shaped** (symmetric), implying the true distribution of exchange rates has **heavier tails** than lognormal.

🟢 **Comment / Application:**

> Two conditions for lognormality that fail in FX markets:
> 1. Volatility of the asset is constant.
> 2. Price changes smoothly with no jumps.
>
> Exchange rates exhibit **jumps** and **stochastic volatility**, causing heavier tails and a higher central peak (leptokurtosis) compared to lognormal.

---

## 20.3 Volatility Smile for Equity Options

For equities, the smile is a **downward-sloping skew**: implied vol decreases as strike increases.

🟢 **Comment / Application:**

> There is a **negative correlation between equity prices and volatility** (leverage effect and crashophobia):
> - As stock prices fall, leverage rises, increasing risk → volatility goes up
> - Demand for downside puts (crash protection) raises their implied vol
>
> This creates the characteristic negative skew observed in equity markets.

---

## 20.4 Alternative Characterizations of the Volatility Smile

Instead of implied vol vs. *strike*, traders plot vs.:
- Moneyness: $K / S_0$
- Log-moneyness: $\ln(K / F_0)$ where $F_0$ is the forward price
- Option delta

---

## 20.5 The Volatility Term Structure and Volatility Surfaces

A **volatility surface** combines the smile with the term structure. Key observation:
- If current vol > long-run average → downward-sloping term structure
- If current vol < long-run average → upward-sloping term structure

Traders interpolate between quoted market prices to build the surface.

---

## 20.6 Minimum Variance Delta

🟡 **Property:**

> The standard BSM delta is not the minimum variance hedge ratio when there is an empirical relationship between implied vol and the underlying price (as in equity markets).
>
> The **minimum variance delta** adds a correction for the smile sensitivity:
>
> $$\Delta_{MV} = \Delta_{BSM} + \nu_{BSM} \cdot \frac{\partial \sigma_{imp}}{\partial S_0}$$
>
> where the second term accounts for the change in implied vol when the stock price changes.

---

## 20.7 The Role of the Model

Traders use BSM as an **interpolation / extrapolation tool**, not a literal belief in lognormality. The volatility surface provides consistent pricing and hedging across all strikes and maturities.

---

## 20.8 When a Single Large Jump Is Anticipated

If a single large jump is anticipated (e.g., before an earnings announcement or political event), the implied distribution becomes **bimodal**, creating a characteristic dip in the middle of the volatility smile.

---

## Summary

| Concept | Key Point |
|---------|-----------|
| Volatility smile | Implied vol varies with strike |
| Volatility surface | Implied vol varies with strike AND maturity |
| FX smile | U-shaped → heavy tails from jumps and stochastic vol |
| Equity skew | Downward sloping → leverage effect + crashophobia |
| Minimum variance delta | Augments BSM delta with smile sensitivity |
