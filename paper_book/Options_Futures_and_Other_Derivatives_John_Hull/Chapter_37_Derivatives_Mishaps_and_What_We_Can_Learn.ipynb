## Overview

This chapter reviews major derivatives disasters and extracts practical lessons for all users of derivatives: companies, financial institutions, and nonfinancial corporations. The cases range from Metallgesellschaft and Orange County to Barings Bank and Long-Term Capital Management.

---

## 37.1 Lessons for All Users of Derivatives

### Big Picture Lessons

**1. Make sure you fully understand the trade**

Many derivatives disasters occurred because decision-makers did not understand the risks they were taking on. Complex structured products should be subjected to rigorous internal review before being entered into.

**2. Segregate the duties of front, middle, and back office**

The trader (front office), risk manager (middle office), and settlement/accounting (back office) functions must be separated. Nick Leeson at Barings controlled both trading and back-office settlement — a critical control failure.

**3. Set limits and monitor them**

Position limits, loss limits, and VaR limits must be set and enforced. Exceeding limits should trigger immediate escalation. Many disasters involved traders who exceeded limits and hid losses.

**4. Do not assume the past predicts the future**

LTCM and other hedge funds relied heavily on historical correlations that broke down during the 1998 Russian crisis. Extreme scenarios must be stress-tested beyond historical data.

**5. Beware of liquidity risk**

Even if a position is theoretically profitable, it may be impossible to exit without moving the market. Liquidity risk interacts dangerously with leverage.

**6. Do not ignore model risk**

Models are simplifications of reality. Mis-specified models can give dramatically wrong prices/risks. Independent model validation is essential.

## 37.2 Lessons for Financial Institutions

### Key Cases

**Barings (1995)**: Nick Leeson lost £827 million trading Nikkei futures and options. Root cause: inadequate separation of front and back office; weak internal controls.

**Metallgesellschaft (1993)**: Lost ~$1.5 billion hedging long-term oil supply contracts with short-term futures. The hedge created massive margin calls in a falling market. Root cause: mismatch between the duration of liabilities (long-term) and hedging instruments (short-term).

**Orange County (1994)**: Robert Citron invested the county's treasury in leveraged interest rate bets. Lost $1.7 billion when rates rose. Root cause: excessive leverage; inadequate governance.

**LTCM (1998)**: Nobel-Prize-backed hedge fund lost $4 billion in 4 months when Russian default triggered a flight to quality. Root cause: excessive leverage; underestimated tail risk and liquidity risk; model assumptions broke down in crisis.

### Institutional Lessons

**1. Enforce limits strictly**: Trading limits must be set conservatively and strictly enforced — exceptions create moral hazard.

**2. Stress-test your models**: VaR models based on normal distributions underestimate crisis losses. Fat tails and correlation breakdowns must be explicitly modeled.

**3. Watch liquidity, not just market risk**: Positions that are profitable on paper may be impossible to unwind at reasonable cost in a crisis.

**4. Require transparency**: All trades should be visible to senior management and risk managers in real time.

**5. Be skeptical of high Sharpe ratios**: Unusually high risk-adjusted returns often indicate hidden risks (e.g., tail risk being "sold" through options writing).

## 37.3 Lessons for Nonfinancial Corporations

### Key Cases

**Procter & Gamble (1994)**: Entered complex leveraged interest rate derivatives without understanding the exposure. Lost $157 million. Root cause: purchased products that were not understood; insufficient oversight.

**Gibson Greetings (1994)**: Similarly purchased highly complex interest rate derivatives. Lost $20 million. Root cause: inadequate understanding of what was purchased.

**Ashanti Goldfields (1999)**: Hedged gold production too aggressively. When gold prices rose, hedge created large losses. Root cause: hedge was too large relative to production.

### Corporate Lessons

**1. Use derivatives only to hedge, not speculate**: Nonfinancial corporations should use derivatives to reduce business risks — not to make directional bets.

**2. Understand exactly what you are buying**: Before entering any derivative transaction, the risk profile must be fully understood by the decision-maker.

**3. Match hedges to exposures**: Hedge ratio should match the actual business exposure — over-hedging creates speculative positions.

**4. Maintain board-level oversight**: The board should approve a derivatives policy that specifies permitted instruments, maximum exposures, and required reporting.

**5. Be cautious with complex structured products**: Structured products are often presented attractively by banks but contain hidden optionality that can generate large losses.

## Summary of Common Themes Across Disasters

| Theme | Example |
|-------|---------|
| Inadequate controls / segregation of duties | Barings (Leeson) |
| Excessive leverage | LTCM, Orange County |
| Model risk / wrong assumptions | LTCM, MGRM |
| Duration mismatch | Metallgesellschaft |
| Failure to understand the product | P&G, Gibson Greetings |
| Correlation breakdown in crises | LTCM |
| Liquidity risk underestimated | LTCM |
| Insufficient board oversight | Orange County, P&G |

The overarching lesson: **derivatives are powerful risk management tools when used properly, but can amplify losses catastrophically when used improperly, excessively, or without understanding.**
