## Overview

This chapter develops the mathematical framework underlying risk-neutral pricing. It generalizes the risk-neutral world to other numeraires, introduces the concept of a martingale measure, and shows how changing the numeraire simplifies the pricing of interest rate derivatives.

---

## 28.1 The Market Price of Risk

For a derivative $f$ dependent on a single source of uncertainty $\theta$:

$$df = \mu_f f\,dt + \sigma_f f\,dz$$

The **market price of risk** $\lambda$ associated with $z$ is defined by:

$$\mu_f - r = \lambda \sigma_f$$

This must hold for all derivatives dependent on $z$, regardless of their maturity or payoff structure.

🟡 **Property (Girsanov's Theorem):**

> **Girsanov's theorem**: choosing a particular market price of risk $\lambda$ is equivalent to defining a **probability measure**. The real-world measure corresponds to the market price of risk consistent with observed expected returns. The risk-neutral measure sets $\lambda = 0$ for all risks, making all assets drift at the risk-free rate $r$.
>
> Under the risk-neutral measure, the derivative price process becomes:
> $$df = rf\,dt + \sigma_f f\,d\tilde{z}$$
> where $d\tilde{z} = dz + \lambda\,dt$ is a Wiener process under the new measure.

🟡 **Property (CAPM interpretation):**

> Under CAPM, risks **correlated with the market portfolio** (systematic risks) command a risk premium; **idiosyncratic risks** (uncorrelated with the market) do not. Thus the market price of risk $\lambda$ is proportional to the correlation of the risk factor with the market return.

---

## 28.2 Several State Variables

When there are $n$ sources of uncertainty $z_1, \ldots, z_n$, each has its own market price of risk $\lambda_i$. The drift of any traded asset satisfies:

$$\mu - r = \sum_{i=1}^{n} \lambda_i \sigma_i$$

This generalizes the single-factor case to multi-dimensional uncertainty.

---

## 28.3 Martingales

🟡 **Definition:**

> A **martingale** is a zero-drift stochastic process:
>
> $$du = \sigma\,dz$$
>
> where $\sigma$ may itself be stochastic. Key property:
>
> $$E[u_T] = u_0$$
>
> The expected value of a martingale at any future time equals its current value. This makes martingales ideal for pricing — the discounted price process of a traded asset should be a martingale under the risk-neutral measure.

---

## 28.4 Alternative Choices for the Numeraire

🟡 **Definition:**

> A **numeraire** $g$ is a security (always positive) used as the reference unit for measuring prices. The ratio $f/g$ (for any tradeable asset $f$) is a martingale under the measure associated with numeraire $g$ (the **equivalent martingale measure**).
>
> Different numeraire choices simplify different pricing problems:
>
> | Numeraire $g$ | Associated Measure | Use Case |
> |---------------|-------------------|----------|
> | Money market account $e^{rt}$ | Risk-neutral measure $\tilde{\mathbb{Q}}$ | Standard derivatives |
> | Zero-coupon bond $P(t,T)$ | $T$-forward measure | Bond options, interest rate caps |
> | Annuity $A$ | Annuity measure / swap measure | Swaptions |

🟡 **Property (Equivalent Martingale Measure Result):**

> When there are no arbitrage opportunities, for any given numeraire $g$, there exists a unique probability measure under which $f/g$ is a martingale for **all** traded assets $f$. This measure is the **equivalent martingale measure** associated with $g$.

🟡 **Definition (Numeraire Ratio):**

> The drift of $v$ in the measure associated with numeraire $w$ is:
>
> $$\alpha_v = \sum_{i=1}^{n} \sigma_{w,i} \sigma_{v,i}$$
>
> or equivalently:
>
> $$\alpha_v = \rho \sigma_v \sigma_w$$
>
> where $\sigma_v$, $\sigma_w$ are total volatilities and $\rho$ is the instantaneous correlation. This is the key formula for computing drift adjustments when changing numeraires.

🟢 **Comment:**

> When the market price of risk $\lambda = 0$ (risk-neutral world), $f$ follows $df = rf\,dt + \sigma_f f\,dz$. The numeraire in this world is the money market account $e^{rt}$. In a world where $\lambda = \sigma_w$ (the volatility of $w$), $f/w$ is a martingale.

🟢 **Comment:**

> The numeraire ratio formula ($\alpha_v = \rho\sigma_v\sigma_w$) appears throughout interest rate modeling. It explains convexity adjustments, timing adjustments, and quanto adjustments — all of which arise from changing the measure.

---

## 28.5 Extension to Several Factors

With multiple factors, the drift adjustment for changing numeraire from $g_1$ to $g_2$ is:

$$\alpha_v = \sum_{i=1}^{n} \sigma_{g_2,i} \sigma_{v,i}$$

The result holds in full generality for multi-factor models.

---

## 28.6 Black's Model Revisited

Black's model (for bond options, caps, swaptions) is shown to be consistent with the martingale framework:

- For bond options: use the zero-coupon bond $P(t,T)$ as numeraire → **T-forward measure** where the forward bond price is a martingale
- For caplets: use the discount bond maturing at the caplet's payment date → forward LIBOR is lognormal under this measure
- For swaptions: use the annuity $A$ as numeraire → forward swap rate is a martingale under the swap measure

---

## 28.7 Option to Exchange One Asset for Another

Margrabe's formula revisited: To price an exchange option, use the asset $S_2$ being given up as the numeraire. Then $S_1/S_2$ is a martingale under the $S_2$-measure, and the exchange option becomes a standard call on $S_1/S_2$ with strike 1.

---

## 28.8 Change of Numeraire

🟡 **Property:**

> The **change of numeraire** technique: to price a derivative in a convenient measure, identify the numeraire that makes the underlying a martingale. Then:
>
> $$f(0) = g(0) \cdot E^g[f(T)/g(T)]$$
>
> where $E^g$ denotes expectation under the $g$-measure.

🟢 **Comment:**

> This result ($f = g \cdot E^g[f/g]$) is the central pricing equation. By choosing $g$ cleverly, complex drift terms disappear. For example, choosing $g = P(0,T)$ eliminates interest rate uncertainty for derivatives paying off at $T$.

## Summary

| Concept | Key Point |
|---------|-----------|
| Market price of risk | $\mu_f - r = \lambda\sigma_f$ for all derivatives |
| Girsanov's theorem | Changing $\lambda$ changes the probability measure |
| Martingale | $E[u_T] = u_0$; zero drift process |
| Numeraire | Reference asset; $f/g$ is martingale under $g$-measure |
| Risk-neutral measure | Numeraire = money market account $e^{rt}$ |
| T-forward measure | Numeraire = $P(t,T)$; forward prices are martingales |
| Swap measure | Numeraire = annuity $A$; swap rates are martingales |
