## Overview

This chapter discusses commodity derivatives — agricultural, metals, and energy — focusing on the unique features of commodity markets (seasonality, mean reversion, convenience yield, storage costs) and the pricing of weather and insurance derivatives.

---

## 35.1 Agricultural Commodities

🟢 **Comment:**

> Agricultural commodity prices (corn, wheat, soybeans, coffee, cotton, etc.) are driven by seasonal supply/demand patterns, weather, and geopolitical events. Storage costs and spoilage play key roles. Unlike financial assets, agricultural commodities cannot be sold short easily, limiting arbitrage.

---

## 35.2 Metals

Metals (gold, silver, copper, aluminum) have well-developed futures markets. Key concepts:

**Cost-of-carry model** for storable metals:

$$F_0 = S_0 e^{(r+u-y)T}$$

where $u$ = storage cost, $y$ = convenience yield, $r$ = risk-free rate. For gold (no convenience yield): $F_0 = S_0 e^{(r+u)T}$.

---

## 35.3 Energy Products

Energy markets (oil, natural gas, electricity) have special characteristics:
- **Seasonality**: electricity and gas prices spike in winter (heating) and summer (cooling)
- **Mean reversion**: commodity prices tend to revert to long-run marginal cost
- **Jumps**: supply disruptions or demand spikes cause price jumps
- **Inelastic demand**: especially for electricity, short-run demand is price-insensitive

🟢 **Application:**

> **Swing option** (take-and-pay option): a contract specifying a minimum and maximum amount of a commodity (e.g., electricity or gas) that must be purchased at a given price per period. The buyer can "swing" their consumption within these bounds. Valuation requires simulation of the commodity price path.

## 35.4 Modeling Commodity Prices

### Geometric Brownian Motion

Simple model: $dS = \mu S\,dt + \sigma S\,dz$ — ignores mean reversion and seasonality.

### Mean-Reverting Model

🟡 **Definition:**

> A more realistic risk-neutral process for commodity prices incorporates **mean reversion**:
>
> $$d\ln S = [\theta(t) - a\ln S]\,dt + \sigma\,dz \quad (35.2)$$
>
> where $a$ = speed of mean reversion and $\theta(t)$ is chosen to fit the futures price curve. This is the log-normal analogue of the Hull–White interest rate model applied to commodity prices.

🟡 **Property (Seasonality Adjustment):**

> Seasonal factors can be estimated from historical spot prices:
> 1. Compute the 12-month centered moving average of spot prices
> 2. Calculate the ratio of monthly spot price to 12-month moving average
> 3. Average these ratios across years to get the seasonal factor for each month
>
> The seasonal factor $s(t)$ is then incorporated into $\theta(t) = \bar{\theta}(t) + s(t)$.

**Two-factor models**: One factor captures the spot price (mean-reverting); a second factor captures the long-run equilibrium level (itself stochastic). This gives more realistic futures curve dynamics.

## 35.5 Weather Derivatives

Weather derivatives (futures, options) have payoffs linked to temperature, precipitation, or other weather variables. Common measure: **Heating Degree Days (HDD)** and **Cooling Degree Days (CDD)**:

$$\text{HDD} = \max(18°C - \bar{T}_{\text{day}}, 0) \quad \text{or} \quad \text{CDD} = \max(\bar{T}_{\text{day}} - 18°C, 0)$$

where $\bar{T}_{\text{day}}$ is the average daily temperature.

---

## 35.6 Insurance Derivatives

Insurance derivatives (CAT bonds, CAT options) have payoffs linked to insurance losses from catastrophes (hurricanes, earthquakes). They allow insurers to transfer risk to capital markets.

---

## 35.7 Pricing Weather and Insurance Derivatives

Since weather and catastrophe risks are largely **uncorrelated with financial markets**, CAPM implies the market price of risk $\lambda \approx 0$, and the real-world probability distribution can be used for pricing:

$$\text{Price} \approx e^{-rT} E^{\mathbb{P}}[\text{Payoff}]$$

Historical data on temperature / catastrophe losses is used to calibrate the payoff distribution.

---

## 35.8 How an Energy Producer Can Hedge Risks

Energy producers face:
- **Price risk**: commodity price falls after production decisions are made
- **Volume risk**: production or demand may differ from expectations

Hedging tools:
- Futures and forwards: lock in prices
- Options: provide downside protection while maintaining upside
- Swing options: allow flexible purchase/sale quantities
