## Overview

Credit risk arises from the possibility that borrowers or counterparties may default. Key topics: default probabilities, recovery rates, risk-neutral vs. real-world probabilities, credit risk in OTC derivatives, default correlation, and credit VaR.

---

## 24.1 Credit Ratings

Rating agencies (Moody's, S&P, Fitch) assign creditworthiness ratings:

| Moody's | S&P/Fitch | Grade |
|---------|-----------|-------|
| Aaa | AAA | Highest (investment grade) |
| Aa | AA | High quality |
| A | A | Upper-medium |
| Baa | BBB | Lower-medium (lowest IG) |
| Ba, B | BB, B | Speculative ("high yield") |
| Caa–C | CCC–C | Highly speculative / near default |

🟡 **Definition (Unconditional Default Probability):**

> From a ratings transition table, the **unconditional default probability** during year $k$ = cumulative default probability by end of year $k$ minus cumulative default probability by end of year $k-1$.

---

## 24.2 Historical Default Probabilities

🟡 **Definition (Hazard Rate / Default Intensity):**

> The **hazard rate** $\lambda(t)$ satisfies: $\lambda(t)\Delta t$ = probability of default between time $t$ and $t + \Delta t$, *given* survival to time $t$.
>
> Survival probability to time $t$: $e^{-\bar{\lambda}t}$, where $\bar{\lambda}$ is the average hazard rate.
>
> **Cumulative default probability**:
> $$Q(t) = 1 - e^{-\bar{\lambda}t}$$
>
> Another name: **default intensity**.

---

## 24.3 Recovery Rates

🟡 **Property:**

> The **recovery rate** = bond's market value shortly after default / face value.
>
> Recovery rates depend on seniority. Senior secured bonds recover more than subordinated bonds. Average for senior unsecured bonds: ~40%.
>
> Recovery rates are **negatively correlated with default rates**: in bad years (many defaults), recovery rates are also low — compounding losses.

---

## 24.4 Estimating Default Probabilities from Bond Spreads

🟡 **Property:**

> For a bond with yield spread $s$ over risk-free and recovery rate $R$:
>
> $$\bar{\lambda} \approx \frac{s}{1 - R}$$
>
> This gives **risk-neutral default probabilities** — used for pricing credit instruments.

---

## 24.5 Comparison of Default Probability Estimates

🟢 **Comment:**

> Risk-neutral default probabilities (from bond spreads) are consistently **much higher** than real-world (historical) default probabilities. Reasons include:
> - **Default risk premium**: Investors require extra return for bearing default risk
> - **Liquidity premium**: Corporate bonds are less liquid than risk-free bonds
> - **Tax considerations**
>
> The gap was especially large during the 2007–08 financial crisis.

🟢 **Comment:**

> Real-world probabilities are used for **expected loss** calculations. Risk-neutral probabilities are used for **pricing** (CVA, CDS valuation). The ratio of risk-neutral to real-world default probability can exceed 5x for investment-grade bonds.

## 24.6 Merton's Model: Equity Prices and Default Probabilities

Equity = call option on the firm's assets with strike = debt face value $D$:

$$E = V N(d_1) - D e^{-rT} N(d_2)$$

$$d_1 = \frac{\ln(V/D) + (r + \sigma_V^2/2)T}{\sigma_V\sqrt{T}}, \quad d_2 = d_1 - \sigma_V\sqrt{T}$$

**Risk-neutral default probability**: $N(-d_2)$

Given observable equity value $E$ and equity volatility $\sigma_E$, solve for asset value $V$ and asset volatility $\sigma_V$ using:
$$\sigma_E E = N(d_1)\sigma_V V$$

## 24.7 Credit Risk in Derivatives Transactions

🟡 **Property (Netting):**

> Under the **ISDA Master Agreement**, if a counterparty defaults, only the **net** value of all transactions counts — not the gross sum of positive exposures. This substantially reduces credit exposure.

🟡 **Property (Margin Period of Risk):**

> The **cure period** (margin period of risk) $c$ = time from when a counterparty stops posting collateral to actual default. Typically 10–20 days. Determines the maximum collateral shortfall.

🟡 **Property (Wrong-Way Risk):**

> **Wrong-way risk** occurs when the probability of counterparty default is positively correlated with the bank's exposure to that counterparty. This is the worst case for credit risk management.

🟡 **Property (Downgrade Trigger):**

> A **downgrade trigger** clause in the Master Agreement allows early termination / collateral posting if the counterparty's credit rating drops below a threshold.

🟡 **Property (Collateral):**

> **Collateral agreements** (CSAs) significantly reduce credit exposure. Collateral can be cash (earning interest) or marketable securities. Collateral management introduces **gap risk** during the margin period of risk.

**CVA (Credit Value Adjustment):**

$$\text{CVA} \approx \sum_i (1-R)\lambda_i\Delta t \cdot E_i$$

where $E_i$ = expected exposure in period $i$, $\lambda_i$ = hazard rate, $R$ = recovery rate.

## 24.8 Default Correlation

🟡 **Definition:**

> **Default correlation** = the tendency for companies to default around the same time, driven by shared macroeconomic exposures.

🟡 **Property (Model Types):**

> - **Reduced form models**: Hazard rates follow correlated stochastic processes driven by common macroeconomic factors
> - **Structural models**: Correlate companies through their asset value processes (Merton-type); a company defaults when assets fall below debt

🟡 **Definition (Gaussian Copula):**

> The most popular practical tool: transforms each company $i$'s time to default into a standard normal $x_i$.
>
> **One-factor model**:
> $$x_i = a_i F + \sqrt{1 - a_i^2} Z_i$$
>
> where $F$ is a common factor (macroeconomic), $Z_i$ is idiosyncratic, and $\text{corr}(x_i, x_j) = a_i a_j$.

---

## 24.9 Credit VaR

🟡 **Property (CreditMetrics):**

> **CreditMetrics**: estimate the distribution of credit losses via Monte Carlo simulation of credit rating migrations.
> 1. Draw correlated samples of normalized asset returns $x_i$ (using Gaussian copula)
> 2. Map each $x_i$ to a credit rating change (or default)
> 3. Compute portfolio loss for each scenario
> 4. VaR = selected percentile of the resulting loss distribution
