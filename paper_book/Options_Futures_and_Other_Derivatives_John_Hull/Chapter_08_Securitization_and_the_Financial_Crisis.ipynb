## Overview

This chapter explains securitization — the process of pooling loans and issuing securities backed by them — and examines how flawed securitization practices contributed to the 2007–08 financial crisis.

---

## 8.1 Securitization

🟡 **Definition (Asset-Backed Security / ABS):**

> In an **ABS**, a portfolio of income-producing assets (e.g., mortgages, credit card receivables) is sold by an originating bank to a **Special Purpose Vehicle (SPV)** — a separate legal entity that issues securities (tranches) backed by the cash flows from the assets.
>
> The SPV issues tranches with different priorities:
> - **Senior tranche**: paid first; highest credit quality; lowest yield
> - **Mezzanine tranche**: paid after senior; intermediate risk/yield
> - **Equity tranche**: paid last (residual); absorbs first losses; highest risk/yield

This structure is a **waterfall** — losses flow from the bottom (equity) upward.

---

## 8.2 The U.S. Housing Market 2000–2007

Key developments:
- Low interest rates post-2001 tech bust → housing demand surge
- Rise of **subprime mortgages**: loans to borrowers with poor credit histories
- **Ninja loans**: No Income, No Job, No Assets
- Home price appreciation made refinancing possible, masking credit quality
- Originate-to-distribute model: banks originated loans but sold them (reducing incentive to screen borrowers carefully)

---

## 8.3 CDOs of ABS

- Subprime mortgages → **ABS** → mezzanine tranches of ABS → **ABS CDO** (CDO of ABS)
- Senior tranches of ABS CDOs were rated AAA but were backed by the worst subprime collateral
- Rating agencies systematically underestimated the correlation of default across mortgages

---

## 8.4 Role of Incentives and Moral Hazard

**Problems in the originate-to-distribute model:**
- Mortgage originators: had little incentive to verify borrower quality (loans were sold)
- Banks: earned fees for packaging and selling securities
- Rating agencies: paid by issuers → conflict of interest
- Investors: relied on ratings without doing independent due diligence

---

## 8.5 The Crisis Unfolds

Timeline:
- **2006**: U.S. home prices peak and begin to decline
- **Early 2007**: Subprime mortgage delinquencies spike
- **August 2007**: Money markets freeze; BNP Paribas suspends redemptions
- **September 2008**: Lehman Brothers fails → global financial crisis

**Why the crisis spread globally:**
- ABS and CDO securities were held by banks worldwide
- Uncertainty about which institutions held "toxic" assets → liquidity crisis
- Interbank lending froze → credit crunch → recession

---

## 8.6 Lessons Learned

1. **Skin in the game**: originators should retain some exposure to the loans they make
2. **Transparency**: complex structured products must have transparent underlying collateral
3. **Rating agency reform**: ratings for structured products should not be produced by agencies paid by issuers
4. **Correlation risk**: ABS ratings failed to account for the high correlation of subprime mortgage defaults during a systemic housing decline
5. **Regulatory capital**: banks held off-balance-sheet ABS vehicles with insufficient capital
