## 14.1 The Markov Property

### 🟡 Definition: Stochastic Process
A variable whose value changes over time in an uncertain way follows a **stochastic process**. They are classified along two axes:
- **Discrete time** vs. **Continuous time** (when changes can occur)
- **Discrete variable** vs. **Continuous variable** (what values can be taken)

### 🟡 Definition: Markov Process
A **Markov process** is a particular type of stochastic process where **only the current value** of a variable is relevant for predicting the future. The past history of the variable and the way the present emerged from the past are irrelevant.

- The probability distribution of the price at any future time is **not dependent** on the particular path followed in the past.

### 🟢 Application: Weak-Form Market Efficiency
The Markov property of stock prices is consistent with the **weak form of market efficiency**, which states that the present price impounds all information contained in a record of past prices.

- If weak-form efficiency were violated, technical analysts could earn above-average returns from historical price charts — there is very little evidence they can.
- **Competition** in the marketplace tends to ensure weak-form efficiency holds: any discovered pattern would be quickly exploited and arbitraged away.

## 14.2 Continuous-Time Stochastic Processes

### 🟡 Property: Variance Additivity for Markov Processes
For a Markov process with current value 10 and annual change $\phi(0, 1)$:
- Change over time $T$: $\phi(0, T)$
- Change over $\Delta t$: $\phi(0, \Delta t)$

> **Key insight:** Variances of changes in successive time periods are **additive**. Standard deviations are **not** additive. Uncertainty (measured by standard deviation) is proportional to $\sqrt{T}$ — the **square root of time**.

---

### 🟡 Definition: Wiener Process (Brownian Motion)
A **Wiener process** is a Markov stochastic process with mean change of zero and variance rate of 1.0 per year. A variable $z$ follows a Wiener process if:

**Property 1.** The change $\Delta z$ during a small period $\Delta t$ is:
$$\Delta z = \epsilon \sqrt{\Delta t}$$
where $\epsilon \sim \phi(0, 1)$ (standard normal).

**Property 2.** The values of $\Delta z$ for any two different short intervals $\Delta t$ are **independent**.

It follows that:
- $E[\Delta z] = 0$
- $\text{Var}(\Delta z) = \Delta t$
- $\text{SD}(\Delta z) = \sqrt{\Delta t}$

Over a longer period $T$:
$$z(T) - z(0) = \sum_{i=1}^{N} \epsilon_i \sqrt{\Delta t}, \quad N = T / \Delta t$$

So $z(T) - z(0) \sim \phi(0, T)$ with standard deviation $\sqrt{T}$.

---

### 🟡 Definition: Generalized Wiener Process
$$dx = a\,dt + b\,dz$$
where $a$ and $b$ are constants.

- **Drift rate** (expected change per unit time): $a$
- **Variance rate** (variance per unit time): $b^2$

In a small interval $\Delta t$:
$$\Delta x = a\,\Delta t + b\,\epsilon\sqrt{\Delta t}$$

| Property | Value |
|---|---|
| $E[\Delta x]$ | $a\,\Delta t$ |
| $\text{SD}(\Delta x)$ | $b\sqrt{\Delta t}$ |
| $\text{Var}(\Delta x)$ | $b^2 \Delta t$ |

Over time $T$: mean change $= aT$, SD of change $= b\sqrt{T}$.

### 🟢 Application: Cash Position Example
A company's cash position (in \$000s) follows a generalized Wiener process with drift $a = 20$/year, variance rate $b^2 = 900$/year, initial cash $= 50$.
- End of 1 year: $\mathcal{N}(70, 30^2)$
- End of 6 months: $\mathcal{N}(60, 21.21^2)$

---

### 🟡 Definition: Itô Process
$$dx = a(x, t)\,dt + b(x, t)\,dz$$

A generalized Wiener process where the drift $a$ and volatility coefficient $b$ are **functions of $x$ and $t$**. Both drift rate and variance rate can change over time. The process is still Markov because the change depends only on the current value of $x$.

## 14.3 The Process for a Stock Price

### 🟡 Definition: Geometric Brownian Motion (GBM)
The most widely used model for stock price behavior:

$$dS = \mu S\,dt + \sigma S\,dz$$

or equivalently:

$$\frac{dS}{S} = \mu\,dt + \sigma\,dz$$

where:
- $\mu$ = expected rate of return (annualized)
- $\sigma$ = volatility of the stock price
- $\sigma^2$ = variance rate

**Why not a simple generalized Wiener process?** Investors require the same **percentage return** regardless of the stock price level. So the expected **drift rate** should be $\mu S$ (proportional to $S$), not a constant.

### 🟡 Discrete-Time Approximation
$$\frac{\Delta S}{S} = \mu\,\Delta t + \sigma\,\epsilon\sqrt{\Delta t}$$

$$\frac{\Delta S}{S} \sim \phi(\mu\,\Delta t,\; \sigma^2\,\Delta t)$$

### 🟢 Application: Monte Carlo Simulation
Simulating GBM over small time steps $\Delta t$:
$$\Delta S = \mu S\,\Delta t + \sigma S\,\epsilon\sqrt{\Delta t}$$

1. Sample $\epsilon$ from $\phi(0, 1)$
2. Compute $\Delta S$
3. Update $S \leftarrow S + \Delta S$
4. Repeat

> Example (Table 14.1): $\mu = 0.15$, $\sigma = 0.30$, weekly steps ($\Delta t = 1/52$), $S_0 = 100$. One simulated path reaches $\$111.54$ after 10 weeks.

## 14.4 The Parameters

### 🟢 Comments on $\mu$ and $\sigma$

**On $\mu$ (expected return):**
- Depends on the risk of the stock and the level of interest rates
- Fortunately, derivative values are (in general) **independent of $\mu$**

**On $\sigma$ (volatility):**
- Critically important for derivative pricing
- Typical values: $0.15$ to $0.60$ (15% to 60% per annum)
- $\sigma\sqrt{\Delta t}$ ≈ standard deviation of the proportional change over $\Delta t$
- $\sigma\sqrt{T}$ ≈ standard deviation of the proportional change over $T$ (rough approximation)
- Can be interpreted as the standard deviation of the continuously compounded return over 1 year

## 14.5 Correlated Processes

### 🟡 Property: Correlated Wiener Processes
For two variables:
$$dx_1 = a_1\,dt + b_1\,dz_1, \quad dx_2 = a_2\,dt + b_2\,dz_2$$

- If **uncorrelated**: sample $\epsilon_1$ and $\epsilon_2$ independently
- If **correlation** $\rho$: sample from a bivariate normal with correlation $\rho$

To generate correlated samples:
$$\epsilon_1 = u, \quad \epsilon_2 = \rho u + \sqrt{1 - \rho^2}\,v$$
where $u, v$ are independent standard normals.

> Generalizes to $n$ correlated variables via multivariate normal sampling (Cholesky decomposition, see Chapter 21).

## 14.6 Itô's Lemma

### 🟡 Theorem: Itô's Lemma
If $x$ follows the Itô process $dx = a(x,t)\,dt + b(x,t)\,dz$, and $G$ is a function of $x$ and $t$, then:

$$\boxed{dG = \left(\frac{\partial G}{\partial x}a + \frac{\partial G}{\partial t} + \frac{1}{2}\frac{\partial^2 G}{\partial x^2}b^2\right)dt + \frac{\partial G}{\partial x}b\,dz}$$

$G$ also follows an Itô process with:
- **Drift rate:** $\frac{\partial G}{\partial x}a + \frac{\partial G}{\partial t} + \frac{1}{2}\frac{\partial^2 G}{\partial x^2}b^2$
- **Variance rate:** $\left(\frac{\partial G}{\partial x}\right)^2 b^2$

> **Key insight:** The Wiener process $dz$ underlying both $x$ and $G$ is **the same** — both are subject to the same source of uncertainty. This is central to the derivation of Black–Scholes–Merton.

### 🟢 Application to Stock Price Derivatives
If $dS = \mu S\,dt + \sigma S\,dz$, then a function $G(S, t)$ follows:

$$dG = \left(\frac{\partial G}{\partial S}\mu S + \frac{\partial G}{\partial t} + \frac{1}{2}\frac{\partial^2 G}{\partial S^2}\sigma^2 S^2\right)dt + \frac{\partial G}{\partial S}\sigma S\,dz$$

### 🟢 Application: Forward Contracts
For $F = Se^{r(T-t)}$:

$$dF = (\mu - r)F\,dt + \sigma F\,dz$$

The forward price follows GBM with **expected growth rate** $\mu - r$ (instead of $\mu$) and the **same volatility** $\sigma$.

## 14.7 The Lognormal Property

### 🟡 Theorem: Log-Normality of Stock Prices
Setting $G = \ln S$ and applying Itô's lemma to $dS = \mu S\,dt + \sigma S\,dz$:

$$d(\ln S) = \left(\mu - \frac{\sigma^2}{2}\right)dt + \sigma\,dz$$

This is a **generalized Wiener process** with constant drift $\mu - \sigma^2/2$ and constant variance rate $\sigma^2$.

Therefore:

$$\boxed{\ln S_T - \ln S_0 \sim \phi\left[\left(\mu - \frac{\sigma^2}{2}\right)T,\; \sigma^2 T\right]}$$

$$\ln S_T \sim \phi\left[\ln S_0 + \left(\mu - \frac{\sigma^2}{2}\right)T,\; \sigma^2 T\right]$$

> Since $\ln S_T$ is normally distributed, $S_T$ has a **lognormal distribution**. The standard deviation of $\ln S_T$ is $\sigma\sqrt{T}$, proportional to the square root of the time horizon.

## 14.8 Fractional Brownian Motion

### 🟡 Definition: Fractional Brownian Motion
A generalization of Brownian motion where:

$$E[(X(t) - X(s))^2] = \sigma^2 (t - s)^{2H}$$

where $H$ is the **Hurst exponent**.

| $H$ Value | Behavior |
|---|---|
| $H = 0.5$ | Standard Brownian motion (Markov) |
| $H > 0.5$ | Positive correlation between successive changes (trending / persistent) |
| $H < 0.5$ | Negative correlation between successive changes (mean-reverting / anti-persistent) |

### 🟡 Properties
- **Non-Markov** (except when $H = 0.5$)
- Correlation between $X(t)$ and $X(s)$: $\;\frac{0.5(t^{2H} + s^{2H} - (t-s)^{2H})}{s^H t^H}$
- Used in **rough volatility models** for derivatives pricing
- As $H$ decreases, the process becomes more noisy

## Summary

| Concept | Key Formula | Type |
|---|---|---|
| Wiener Process | $\Delta z = \epsilon\sqrt{\Delta t}$, $\;\epsilon \sim \phi(0,1)$ | Building block |
| Generalized Wiener | $dx = a\,dt + b\,dz$ | Constant drift & variance |
| Itô Process | $dx = a(x,t)\,dt + b(x,t)\,dz$ | Variable drift & variance |
| GBM | $dS/S = \mu\,dt + \sigma\,dz$ | Stock price model |
| Itô's Lemma | $dG = (G_x a + G_t + \frac{1}{2}G_{xx}b^2)dt + G_x b\,dz$ | Core theorem |
| Lognormal Property | $\ln S_T \sim \mathcal{N}(\ln S_0 + (\mu - \sigma^2/2)T,\; \sigma^2 T)$ | Key result |
| Fractional BM | $E[(X(t)-X(s))^2] = \sigma^2(t-s)^{2H}$ | Generalization |

### Key Takeaways
1. **Wiener process** is the fundamental building block; all other processes build on it
2. **Variances** are additive over time; standard deviations grow with $\sqrt{T}$
3. **Itô's Lemma** is the chain rule of stochastic calculus — with an extra $\frac{1}{2}G_{xx}b^2$ term due to the $\Delta z^2 \sim \Delta t$ property
4. **GBM** implies stock prices are **lognormally distributed**
5. Both the stock and any derivative of it share the **same Wiener process** $dz$

## Appendix: Non-Rigorous Derivation of Itô's Lemma

### 🟡 Key Insight: Why the Second-Order Term Survives

In ordinary calculus, Taylor expansion gives $dG = G_x dx + G_y dy$ (first-order terms only, since $\Delta x^2 \to 0$).

For an Itô process, $\Delta x = a\Delta t + b\epsilon\sqrt{\Delta t}$, so:

$$\Delta x^2 = b^2 \epsilon^2 \Delta t + \text{higher order terms}$$

Since $E[\epsilon^2] = 1$, the term $\epsilon^2 \Delta t$ is **of order $\Delta t$** (not $\Delta t^2$) and **cannot be ignored**. This gives the extra $\frac{1}{2}G_{xx}b^2\,dt$ term in Itô's lemma.

### 🟡 Multi-variable Extension
For $G(x_1, \ldots, x_n, t)$ with $dx_i = a_i\,dt + b_i\,dz_i$:

$$dG = \left(\sum_i \frac{\partial G}{\partial x_i}a_i + \frac{\partial G}{\partial t} + \frac{1}{2}\sum_i \sum_j \frac{\partial^2 G}{\partial x_i \partial x_j}b_i b_j \rho_{ij}\right)dt + \sum_i \frac{\partial G}{\partial x_i}b_i\,dz_i$$

where $\rho_{ij}$ is the correlation between $dz_i$ and $dz_j$.
