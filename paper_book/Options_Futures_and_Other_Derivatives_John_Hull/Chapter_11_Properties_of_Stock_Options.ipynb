## Overview

This chapter derives important theoretical properties of option prices: the six factors that affect option prices, upper and lower bounds, put–call parity, and when early exercise of American options is optimal.

---

## 11.1 Factors Affecting Option Prices

🟡 **Property:**

> Six factors affect the price of a European stock option:
>
> | Factor | Call Price | Put Price |
> |--------|-----------|-----------|
> | Current stock price $S_0$ ↑ | ↑ | ↓ |
> | Strike price $K$ ↑ | ↓ | ↑ |
> | Time to expiration $T$ ↑ | ↑ (American; usually ↑ European) | ↑ (American; usually ↑ European) |
> | Volatility $\sigma$ ↑ | ↑ | ↑ |
> | Risk-free rate $r$ ↑ | ↑ | ↓ |
> | Dividends ↑ | ↓ | ↑ |

🟡 **Property (Volatility Effect):**

> Volatility = measure of uncertainty about future stock price movements. As volatility increases, the probability of both very high and very low outcomes increases. Since the option holder benefits from extreme moves (and is protected from the bad side by limited downside), both calls and puts **increase in value** with higher volatility.

🟡 **Property (Interest Rate Effect):**

> Higher interest rates:
> 1. Increase the expected return investors require on the stock (making the stock grow faster in a risk-neutral world)
> 2. Decrease the present value of future cash flows received by the option holder
>
> Effect (1) dominates for calls (increases their value); effect (2) dominates for puts (decreases their value).

## 11.2 Upper and Lower Bounds for Option Prices

### Upper Bounds

- Call: $c \leq S_0$ (a call cannot be worth more than the stock)
- Put: $p \leq K e^{-rT}$ (a put cannot pay more than $K$, in PV terms)

### Lower Bounds (European, No Dividends)

- Call: $c \geq \max(S_0 - K e^{-rT}, 0)$
- Put: $p \geq \max(K e^{-rT} - S_0, 0)$

**Proof for call lower bound**: Create a portfolio: long call + invest $K e^{-rT}$ at risk-free rate. At expiry, portfolio value = $\max(S_T, K) \geq K$. Today, portfolio value $\geq K e^{-rT}$. Since portfolio = call + $K e^{-rT}$, we get $c \geq S_0 - K e^{-rT}$ (after subtracting today's stock price via a separate arbitrage argument).

---

## 11.3 Put–Call Parity

🟡 **Property (Put–Call Parity for European Options):**

> For European options on a **non-dividend-paying** stock:
>
> $$c + K e^{-rT} = p + S_0$$
>
> Proof: Consider two portfolios:
> - Portfolio A: long call + invest $Ke^{-rT}$ at risk-free rate
> - Portfolio B: long put + long stock
>
> Both have payoff $\max(S_T, K)$ at expiry → same value today. QED.

With known dividends (PV = $I$):

$$c + K e^{-rT} = p + S_0 - I$$

With continuous dividend yield $q$:

$$c + K e^{-rT} = p + S_0 e^{-qT}$$

Put–call parity is model-free — it holds as long as markets are complete and there is no arbitrage.

## 11.4 Early Exercise of American Options

### American Call on a Non-Dividend-Paying Stock

🟡 **Property:**

> **It is never optimal to exercise early** an American call option on a non-dividend-paying stock. Therefore, $C = c$ (American call = European call price).
>
> Two reasons:
> 1. **Insurance value**: Holding the call instead of the stock provides downside protection. Early exercise gives up this insurance.
> 2. **Time value of money**: Exercising early pays $K$ immediately rather than at $T$ — you lose the interest on $K$ for the remaining period.
>
> → Always better to sell the call than to exercise it early.

🟢 **Comment:**

> These two reasons together confirm the result formally. The only time early exercise of an American call can be optimal is when **dividends** are paid. Just before an ex-dividend date, it may be optimal to exercise to capture the dividend.

### American Put on a Non-Dividend-Paying Stock

It **can** be optimal to exercise an American put early. When the stock price is very low, the put is deep in-the-money and the interest earned on $K$ (from early exercise) may outweigh the insurance value of waiting.

Therefore: $P \geq p$ (American put > European put).

---

## 11.5 Effect of Dividends

- Dividends reduce the stock price on the ex-dividend date → lower calls, higher puts
- For **American calls**: early exercise just before an ex-dividend date may be optimal to capture the dividend
- For **European options**, dividends reduce the lower bound: $c \geq S_0 - I - K e^{-rT}$

---

## Summary

| Property | Result |
|---------|--------|
| Put–call parity (no dividends) | $c + Ke^{-rT} = p + S_0$ |
| Call lower bound | $c \geq \max(S_0 - Ke^{-rT}, 0)$ |
| Put lower bound | $p \geq \max(Ke^{-rT} - S_0, 0)$ |
| American call (no dividends) | Never exercise early; $C = c$ |
| American put | May exercise early; $P \geq p$ |
| Volatility effect | Both calls and puts increase with $\sigma$ |
