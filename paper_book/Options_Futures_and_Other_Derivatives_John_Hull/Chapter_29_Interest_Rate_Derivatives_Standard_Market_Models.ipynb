## Overview

This chapter presents Black's model — the market standard — for pricing three major interest rate derivatives: bond options, interest rate caps/floors, and swaptions. All three use the same lognormal assumption but applied to different underlying rates.

---

## 29.1 Bond Options

### Embedded Bond Options

🟡 **Definition:**

> **Callable bond**: issuer has the right to buy back the bond at a predetermined price. The bond embeds a **call option for the issuer** (and a put option for the bondholder), so callable bonds offer higher yields than non-callable bonds to compensate.

🟡 **Definition:**

> **Puttable bond**: holder has the right to demand early redemption at a predetermined price. The holder has purchased a **put option**, so puttable bonds offer lower yields (the holder pays for the option by accepting a lower yield).

### Black's Model for Bond Options

🟡 **Property (Clean vs. Dirty Price):**

> The **quoted (clean) price** of a bond excludes accrued interest. The **cash (dirty) price** = clean price + accrued interest. For bond options, the strike $K$ should be set to $K_{\text{clean}} + \text{accrued interest at option expiry}$.

**Black's model for a European call on a bond:**

$$c = P(0,T)[F_B N(d_1) - K N(d_2)]$$

$$d_1 = \frac{\ln(F_B/K) + \sigma_B^2 T/2}{\sigma_B\sqrt{T}}, \quad d_2 = d_1 - \sigma_B\sqrt{T}$$

where $F_B$ = forward bond price, $\sigma_B$ = bond price volatility, $T$ = option expiry.

The **T-forward measure** (numeraire = $P(0,T)$) makes the forward bond price a martingale, justifying the lognormal assumption.

## 29.2 Interest Rate Caps and Floors

### Caps

🟡 **Definition:**

> An **interest rate cap** is a series of call options on interest rates (caplets). It guarantees a maximum interest rate on a floating-rate loan.
>
> For a floating rate resetting at $t_1, t_2, \ldots, t_{n-1}$ with reference rate $R_k$ and cap rate $R_K$, the caplet payoff at time $t_{k+1}$ is:
>
> $$L\,\delta_k \max(R_k - R_K, 0)$$
>
> where $L$ = principal, $\delta_k$ = accrual fraction for period $[t_k, t_{k+1}]$.

### Floors

🟡 **Definition:**

> An **interest rate floor** is a series of put options on interest rates (floorlets). It guarantees a minimum return on a floating-rate investment.
>
> $$\text{Floorlet payoff at } t_{k+1}: L\,\delta_k\max(R_K - R_k, 0)$$

🟡 **Definition:**

> An **interest rate collar** = long cap + short floor (or vice versa), designed to keep the floating rate within a band $[R_{\text{floor}}, R_{\text{cap}}]$.

### Black's Model for Caplets

Each caplet is priced as a call option on the forward rate $F_k$ for period $[t_k, t_{k+1}]$:

$$\text{Caplet}_k = L\,\delta_k P(0, t_{k+1})[F_k N(d_1) - R_K N(d_2)]$$

$$d_1 = \frac{\ln(F_k/R_K) + \sigma_k^2 t_k/2}{\sigma_k\sqrt{t_k}}, \quad d_2 = d_1 - \sigma_k\sqrt{t_k}$$

The numeraire is the bond $P(0, t_{k+1})$, making the forward rate $F_k$ a martingale under the $(t_{k+1})$-forward measure.

🟡 **Property (Spot vs. Flat Volatility):**

> - **Spot volatilities**: Different $\sigma_k$ for each caplet — captures the volatility term structure
> - **Flat volatility**: Single $\sigma$ for all caplets in a cap — market convention for quoting caps
>
> Given flat volatilities for different maturities, spot volatilities can be bootstrapped.

🟡 **Property (Shifted Lognormal Model):**

> The standard lognormal cap model fails when rates are near or below zero. The **shifted lognormal** fix assumes $R + a$ is lognormal (where $a > 0$ is a shift), allowing the model to handle negative rates. For example, with $a = 3\%$, the model can accommodate rates as low as $-3\%$.

## 29.3 European Swap Options (Swaptions)

🟡 **Definition:**

> A **swaption** is an option to enter into an interest rate swap at a future date at a specified **swap rate** (the strike). The holder can enter as fixed-rate payer (payer swaption) or fixed-rate receiver (receiver swaption).

### Black's Model for Swaptions

**Payer swaption** (right to pay fixed rate $R_K$, receive floating):

$$\text{Swaption} = A(0)[s_0 N(d_1) - R_K N(d_2)]$$

$$d_1 = \frac{\ln(s_0/R_K) + \sigma_s^2 T/2}{\sigma_s\sqrt{T}}, \quad d_2 = d_1 - \sigma_s\sqrt{T}$$

where:
- $s_0$ = current forward swap rate
- $\sigma_s$ = swap rate volatility
- $A(0) = \sum_{i=1}^{n} \delta_i P(0, t_i)$ = present value of annuity (the numeraire)

🟢 **Comment:**

> The swaption formula is internally consistent using the **annuity as numeraire** (swap measure):
> 1. Current value of any security = current annuity value × expected security value / annuity value
> 2. Expected swap rate under the annuity measure = forward swap rate
>
> This is the theoretical foundation justifying Black's swaption formula.

## 29.4 Hedging Interest Rate Derivatives

The key Greeks for interest rate derivatives:

- **Delta** ($\Delta$): sensitivity to parallel shift in yield curve
- **DV01** (dollar value of 1bp): change in value for 1 bps parallel shift
- **Gamma** ($\Gamma$): second-order rate sensitivity
- **Vega**: sensitivity to changes in interest rate volatility

Hedging is typically done using liquid vanilla instruments: FRAs, interest rate swaps, and liquid cap/swaption quotes.
