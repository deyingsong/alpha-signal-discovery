# Chapter 30: Convexity, Timing, and Quanto Adjustments

## Overview

This chapter explains three important adjustments that arise when pricing non-standard interest rate and cross-currency derivatives: convexity adjustments (from the nonlinear price-yield relationship), timing adjustments (from early payment relative to the natural payment date), and quanto adjustments (from payoffs in a different currency than the underlying).

---

## 30.1 Convexity Adjustments

### The Bond Convexity Adjustment

🟡 **Definition:**

> The **convexity adjustment** for a bond yield arises because bond price is a **convex** (nonlinear) function of yield. The expected future bond yield $E[y_T]$ is **not** equal to the forward yield $y_F$:
>
> $$E[y_T] \approx y_F - \frac{1}{2}y_F^2\,\sigma_y^2\,T\,\frac{G''(y_F)}{G'(y_F)}$$
>
> where $G(y)$ = bond price as a function of yield. The **convexity adjustment** = $E[y_T] - y_F > 0$ (the expected yield is above the forward yield due to convexity).

### Convexity Adjustment for CMS Rates

🟡 **Property:**

> A **constant maturity swap (CMS)** pays a swap rate of maturity $T^*$ (e.g., 10-year swap rate) at a payment date $T < T^*$.
>
> The convexity adjustment to the forward swap rate is:
>
> $$E[R_{\text{swap}}] \approx R_F + \frac{R_F^2 G'(R_F)}{-G''(R_F)} \cdot \sigma_R^2 T$$
>
> where the adjustment is positive because swap rates also have a convex relationship with bond prices.

## 30.2 Timing Adjustments

🟡 **Property:**

> A **timing adjustment** arises when a rate determined at time $T$ is paid at time $T^*$ rather than at its natural payment date.
>
> Let $R_F$ = forward rate for period $[T, T^*]$, $\sigma_R$ = volatility of $R_F$, and $m$ = compounding frequency. The adjusted expected rate at payment time $T^*$:
>
> $$E[R] \approx R_F + \frac{R_F^2 \sigma_R^2 T (T^* - T)}{1 + R_F/m}$$

🟡 **Property (General Result):**

> The relationship between $W$ (bond price for the delay period) and $R_F$ (forward interest rate compounded $m$ times per year):
>
> $$W = \frac{1}{(1 + R_F/m)^{m(T^*-T)}}$$
>
> so that $\sigma_W = -\sigma_R R_F (T^* - T)/(1 + R_F/m)$.
>
> This volatility relationship drives the timing adjustment through the numeraire ratio formula.

## 30.3 Quantos

🟡 **Definition:**

> A **quanto** (cross-currency derivative) has a payoff defined in one currency but paid in another. Example: an option on the Nikkei index paid in USD rather than JPY.
>
> **Quanto adjustment**: when the underlying is in foreign currency and payment is in domestic currency, the drift of the foreign variable must be adjusted:
>
> $$\text{Adjusted drift} = \text{usual drift} - \rho\,\sigma_S\,\sigma_X$$
>
> where $\sigma_S$ = volatility of the foreign asset, $\sigma_X$ = volatility of the exchange rate (foreign per domestic), and $\rho$ = correlation between the foreign asset and the exchange rate.

The quanto adjustment arises from the change of numeraire (from foreign to domestic money market account). Under the domestic risk-neutral measure, a foreign asset earns:

$$r_f - \rho\,\sigma_S\,\sigma_X$$

where $r_f$ is the foreign risk-free rate and $\rho\sigma_S\sigma_X$ is the quanto adjustment.

---

## Summary

| Adjustment | Source | Direction | Formula |
|------------|--------|-----------|---------|
| Convexity | Nonlinear price-yield relationship | Positive (expected rate > forward rate) | $\approx -\frac{1}{2}y_F^2\sigma_y^2 T G''/G'$ |
| Timing | Payment date differs from natural date | Positive | $\approx R_F \sigma_R^2 T (T^*-T)/(1+R_F/m)$ |
| Quanto | Cross-currency payoff | Depends on correlation | $-\rho\sigma_S\sigma_X$ subtracted from drift |

All three adjustments arise naturally from the **change of numeraire / measure** framework developed in Chapter 28.
