## Overview

Real options apply financial options theory to the valuation of investment decisions. Unlike financial options, the "underlying assets" are real projects or assets. This chapter extends risk-neutral valuation to real assets and describes the main types of embedded real options.

---

## 36.1 Capital Investment Appraisal

🟡 **Definition:**

> The **Net Present Value (NPV)** approach discounts expected future incremental cash flows at a **risk-adjusted discount rate** (RADR) that reflects the systematic risk of the project:
>
> $$\text{NPV} = \sum_t \frac{E[\text{Cash flow}_t]}{(1 + \text{RADR})^t} - \text{Initial investment}$$
>
> Accept if NPV > 0.

🟢 **Comment:**

> The traditional NPV approach has two weaknesses:
> 1. **Ignores embedded options**: many projects contain options to expand, abandon, or delay. These add value that NPV misses.
> 2. **Discount rate estimation**: it is difficult to estimate the correct RADR for complex projects with changing risk profiles.
>
> Consider a company evaluating a new plant: the option to **abandon** the project (like a put option) and the option to **expand** (like a call option) can be worth significant amounts — often ignored by traditional NPV.

## 36.2 Extension of the Risk-Neutral Valuation Framework

🟢 **Comment:**

> The risk-neutral valuation principle applies to real assets just as it does to financial assets. The key is that the **expected growth rate** of the underlying real asset must be set to the risk-free rate **minus** any convenience yield / dividend yield in the risk-neutral world:
>
> $$\text{Risk-neutral drift of underlying} = r - \delta$$
>
> where $\delta$ is the asset's "dividend yield" (e.g., revenue generated by the project, or the convenience yield of a commodity). This parallels the treatment of dividend-paying stocks in Chapter 15.

---

## 36.3 Estimating the Market Price of Risk

The market price of risk $\lambda$ for a real asset variable $\theta$:

$$\lambda = \frac{m - r}{\sigma}$$

where $m$ = expected return on $\theta$, $r$ = risk-free rate, $\sigma$ = volatility of $\theta$.

For variables **correlated with the market portfolio**, $\lambda$ can be estimated using CAPM:

$$m - r = \rho_{\theta,M} \cdot (m_M - r) \cdot \frac{\sigma}{\sigma_M}$$

where $\rho_{\theta,M}$ = correlation between $\theta$ and the market return.

For variables **uncorrelated with the market** (e.g., weather), $\lambda = 0$ and the real-world distribution can be used directly.

---

## 36.4 Application to Business Valuation

The risk-neutral valuation framework can value entire businesses by modeling the underlying revenue/cash flow process. Key inputs:
- Risk-neutral drift of revenue
- Volatility of revenue
- Discount rate = risk-free rate

This converts the complex business valuation problem into a derivative pricing problem.

## 36.5 Evaluating Options in an Investment Opportunity

### Types of Real Options

🟡 **Definition (Abandonment Option):**

> The option to **sell or close down** a project when it performs poorly. This is an **American put option** on the project's value with:
> - Strike = liquidation value of the project (minus closing costs)
> - Mitigates downside risk; increases initial project valuation

🟡 **Definition (Expansion Option):**

> The option to **increase the scale of a project** if it performs well. This is an **American call option** on the value of the additional capacity with:
> - Strike = cost of expansion
> - Increases value when the project has a high-return phase

🟡 **Definition (Contraction Option):**

> The option to **reduce the scale** of a project. This is an **American put option** on the value of the lost capacity with:
> - Strike = present value of future savings from scaling back

🟡 **Definition (Option to Defer):**

> The option to **postpone** an investment decision. Waiting has value when uncertainty is high — information may be revealed that changes the investment decision. This is like an American call option where exercise = starting the project.

---

## Valuation Methodology

For simple real options: BSM or binomial tree (treating the project value as the underlying asset).

For complex, path-dependent real options: Monte Carlo simulation under the risk-neutral measure.

---

## Summary

| Real Option Type | Financial Analogy | Exercise Condition |
|-----------------|-------------------|-------------------|
| Abandonment option | American put on project value | Liquidation value > continuation value |
| Expansion option | American call on additional capacity | NPV of expansion > expansion cost |
| Contraction option | American put on foregone capacity | Savings > opportunity cost |
| Deferral option | American call on project NPV | Optimal timing to start |

**Key principle**: Real options add value over the simple NPV approach whenever management has flexibility to respond to new information.
