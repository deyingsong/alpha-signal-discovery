## Overview

No-arbitrage models extend equilibrium models by incorporating a **time-dependent parameter** that allows an exact fit to the current term structure. This chapter covers the Hull–White, Black–Derman–Toy, and Black–Karasinski models, along with interest rate tree construction and calibration.

---

## 32.1 Extensions of Equilibrium Models

### Hull–White (Extended Vasicek) Model

🟡 **Definition:**

> $$dr = [\theta(t) - ar]\,dt + \sigma\,dz \quad \text{(32.4)}$$
>
> where $\theta(t)$ is chosen to fit the initial term structure exactly. Equivalently:
>
> $$dr = a\left[\frac{\theta(t)}{a} - r\right]\,dt + \sigma\,dz$$
>
> - $a$ = mean reversion speed (constant)
> - $\sigma$ = volatility (constant)
> - $\theta(t)$ = time-dependent reversion level calibrated to the yield curve

$\theta(t)$ is determined analytically from the initial forward rate curve:

$$\theta(t) = F_t(0,t) + aF(0,t) + \frac{\sigma^2}{2a}(1 - e^{-2at})$$

where $F(0,t) = -\partial \ln P(0,t)/\partial t$ is the initial instantaneous forward rate.

**Bond and bond option prices**: Analytical formulas exist (similar to Vasicek but with $\theta(t)$).

### Black–Derman–Toy (BDT) Model

🟡 **Definition:**

> The BDT model uses a **recombining binomial tree** for $\ln r$:
>
> $$d\ln r = \left[\theta(t) - \frac{\sigma'(t)}{\sigma(t)}\ln r\right]\,dt + \sigma(t)\,dz$$
>
> where the reversion rate is $a(t) = -\sigma'(t)/\sigma(t)$. If $\sigma$ is constant, the model has **no mean reversion**.

### Black–Karasinski (BK) Model

🟡 **Definition:**

> $$d\ln r = [\theta(t) - a(t)\ln r]\,dt + \sigma(t)\,dz$$
>
> The most general log-normal short rate model. The **reversion rate** $a(t)$ and **volatility** $\sigma(t)$ can be specified independently (unlike BDT where they are linked). This ensures $r > 0$ always.

## 32.2 Options on Bonds

Hull–White provides analytical formulas for European bond options — similar in structure to Jamshidian's decomposition: an option on a coupon bond can be decomposed into options on zero-coupon bonds.

---

## 32.3 Volatility Structures

- **Short-rate volatility** ($\sigma$): governs the volatility of the short end of the yield curve
- **Long-rate volatility**: governed by $\sigma/a$ (for Hull–White, this is the long-run standard deviation of rates)
- **Term structure of volatility**: the model's volatility of forward rates as a function of maturity

---

## 32.4 Interest Rate Trees

A trinomial tree is typically used for Hull–White:

At each node, three branches: up ($+\Delta r$), middle (no change), down ($-\Delta r$), with risk-neutral probabilities chosen to match the model's drift and variance.

---

## 32.5 A General Tree-Building Procedure (Hull–White Trinomial Tree)

**Two-stage construction:**

**Stage 1**: Build a tree for $r^* = r - F(0,t)$ (deviations from the forward rate) with zero initial value. This tree is symmetric and recombining.

**Stage 2**: Displace each $r^*$ node by $F(0,t) + \alpha(t)$ where $\alpha(t)$ is chosen period-by-period to match the initial term structure (zero-coupon bond prices).

The displacement $\alpha(t)$ is found by solving backward through the tree at each time step.

## 32.6 Calibration

🟡 **Property:**

> **Calibration**: the process of determining model parameters from observed market prices of actively traded instruments (the "calibrating instruments").
>
> **Goodness-of-fit** measure: minimize the sum of squared differences between model and market prices (or implied volatilities) for the calibrating instruments.
>
> Calibrating instruments for interest rate models: liquid caps and swaptions.

---

## 32.7 Hedging Using a One-Factor Model

🟡 **Property (Outside-Model Hedging):**

> One-factor models can only hedge against **parallel shifts** in the yield curve. In practice, traders also hedge against curve twists and other movements not captured by the model.
>
> **Outside-model hedging**: taking account of yield curve changes that cannot occur under the model (e.g., non-parallel shifts in a one-factor model). This is standard practice and provides robustness against model misspecification.

🟢 **Comment:**

> For pricing term-structure derivatives requiring a full yield-curve model (e.g., when two yield curves are involved — such as risk-free OIS and a risky LIBOR curve), one approach is to model the risk-free short rate with Hull–White and treat spreads between curves as separate (possibly stochastic) processes.
